```python
# ============================================================================
# CELL 2: Configuration and LLM transport
# ============================================================================
import os, time, requests

OPENROUTER_URL = "https://openrouter.ai/api/v1/chat/completions"
from dotenv import load_dotenv

assert load_dotenv("Key.env"), "Key.env not found — check it's beside the notebook"
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
assert OPENROUTER_API_KEY, "OPENROUTER_API_KEY not set in Key.env"

MODEL = "nvidia/nemotron-3-ultra-550b-a55b:free"

# Whether the model may do arithmetic, or must quote figures exactly as supplied.
# False during development: a weak model's math errors and its grounding failures look
# identical in the output, and only one of those is what we're testing.
ALLOW_MODEL_ARITHMETIC = False

LLM_AVAILABLE = False
HEADERS  = {"Authorization": f"Bearer {OPENROUTER_API_KEY}",
            "Content-Type": "application/json"}
RETRYABLE = {408, 429, 500, 502, 503, 504}


class _Retry(Exception):
    def __init__(self, code, msg):
        self.code, self.msg = code, msg
        super().__init__(f"{code}: {msg}")


def _unwrap(r):
    """Extract content, or raise. OpenRouter returns 200 with an error body when the
    upstream provider fails — trusting the status line gives a KeyError three frames down."""
    payload = r.json()
    if "error" in payload:
        err, code = payload["error"], payload["error"].get("code", r.status_code)
        if code in RETRYABLE:
            raise _Retry(code, err.get("message", ""))
        raise RuntimeError(f"provider error {code}: {err.get('message', '')[:200]}")
    if not payload.get("choices"):
        raise RuntimeError(f"no choices: {str(payload)[:300]}")

    u = payload.get("usage", {})
    reasoning = (u.get("completion_tokens_details") or {}).get("reasoning_tokens", 0)
    print(f"  tokens: {u.get('completion_tokens', '?')} completion"
          f"{f' ({reasoning} reasoning)' if reasoning else ''}"
          f" | finish: {payload['choices'][0].get('finish_reason')}")

    # A truncated reply is not a shorter reply — it cuts mid-figure, and the checker then
    # flags the fragment as an unsupported number. Fail loudly rather than return it.
    if payload["choices"][0].get("finish_reason") == "length":
        raise RuntimeError(
            f"truncated at max_tokens ({u.get('completion_tokens')} used, "
            f"{reasoning} of them reasoning) — raise max_tokens or use a non-reasoning model")

    return payload["choices"][0]["message"]["content"]


def call_llm(system, user, max_tokens=2000, temperature=0.0, retries=3):
    """Single transport for every LLM call. `system` is a parameter, not a constant —
    a hardcoded persona leaves the citation rules nowhere to live."""
    for attempt in range(retries):
        try:
            r = requests.post(OPENROUTER_URL, headers=HEADERS, timeout=120,
                              json={"model": MODEL, "temperature": temperature,
                                    "max_tokens": max_tokens,
                                    "reasoning": {"enabled": False},
                                    "messages": [{"role": "system", "content": system},
                                                 {"role": "user", "content": user}]})
            if r.status_code in RETRYABLE:
                raise _Retry(r.status_code, r.text[:200])
            return _unwrap(r)
        except _Retry as e:
            wait = 15 * (attempt + 1)
            print(f"  retryable ({e.code}), waiting {wait}s — {e.msg[:120]}")
            time.sleep(wait)
    raise RuntimeError(f"LLM unavailable after {retries} attempts")


def check_llm():
    """Verify the endpoint. Sets LLM_AVAILABLE; never raises."""
    global LLM_AVAILABLE
    try:
        # 200, not 5: a reasoning model spends tokens before emitting anything, so a
        # tight ceiling truncates the probe rather than testing the connection.
        call_llm("Reply with exactly: ok", "ping", max_tokens=200, retries=1)
        LLM_AVAILABLE = True
        print(f"LLM ready: {MODEL} | model arithmetic: "
              f"{'permitted' if ALLOW_MODEL_ARITHMETIC else 'disabled'}")
    except Exception as e:
        LLM_AVAILABLE = False
        print(f"LLM unavailable — {type(e).__name__}: {str(e)[:160]}")
        print("The system will work without the LLM and answer from the data directly.")
    return LLM_AVAILABLE

    

check_llm()
```

```python
# ============================================================================
# CELL 4: Data layer — load and classify
# ============================================================================
import pandas as pd
from pathlib import Path

CSV_DIR  = Path("CSVs")
PNL_CSV  = CSV_DIR / "P&L_2022_2025_addbacks.csv"
TAX_CSV  = CSV_DIR / "Taxes_addback_analysis_all.csv"
PROD_CSV = CSV_DIR / "productivity_2022_2025_all.csv"

# The six P&L add-back categories, in the order method §4 lists them. Named once because
# three separate things need the same list: parsing, summing, and reporting.
PNL_ADDBACKS = ["Automobile", "Contributions", "Legal Fees", "Miscellaneous",
                "Travel & Entertainment", "Owner Excess Compensation"]

# Non-individual rows in the productivity export. Method §5.5: classify by record type
# before computing anything per-provider, or a headcount counts an entity row as a
# clinician. Classified in the loader so downstream code cannot forget to.
RECORD_TYPES = {
    "Roselle Dental Center": "entity",       # practice-entity row, no production
    "IC":                    "contractor",   # independent contractor dentistry, aggregated
    "Hyg Hyg":               "aggregate",    # non-attributed hygiene production
}


def parse_currency(s):
    """'$3,750.00 ' -> 3750.0 | '($6,877.36)' -> -6877.36 | 'NA' -> NaN

    Accounting parentheses mean negative. NA stays NaN rather than becoming 0: a missing
    add-back and a zero add-back are different facts, and collapsing them hides which
    years a category was absent from.
    """
    return pd.to_numeric(
        s.astype(str)
         .str.strip()
         .str.replace(r'[$,]', '', regex=True)
         .str.replace(r'^\((.*)\)$', r'-\1', regex=True)
         .replace({'NA': None, 'N/A': None, 'na': None, '': None}),
        errors="coerce")


def load_pnl(path=PNL_CSV):
    df = pd.read_csv(path)
    df.columns = [c.strip() for c in df.columns]
    for col in df.columns:
        if col != "Year":
            df[col] = parse_currency(df[col])
    df["Year"] = df["Year"].astype(int)
    return df.set_index("Year").sort_index()


def load_tax(path=TAX_CSV):
    df = pd.read_csv(path)
    df.columns = [c.strip() for c in df.columns]
    for col in df.columns:
        if col != "Year":
            df[col] = parse_currency(df[col])
    df["Year"] = df["Year"].astype(int)
    df["Tax_Addbacks"] = df["Depreciation"] + df["Section179"]
    return df.set_index("Year").sort_index()


def load_productivity(path=PROD_CSV):
    """Load the productivity export and tag every row with its record type."""
    df = pd.read_csv(path)
    df.columns = [c.strip() for c in df.columns]
    df["record_type"] = df["provider_name"].map(RECORD_TYPES).fillna("individual")
    df["year"] = df["year"].astype(int)
    return df


pnl  = load_pnl()
tax  = load_tax()
prod = load_productivity()

print(f"P&L          {len(pnl)} years {list(pnl.index)}")
print(f"Tax          {len(tax)} years {list(tax.index)}  (no FY2025 return)")
print(f"Productivity {len(prod)} rows, {prod['year'].nunique()} years")
print()
print("Provider record types:")
for rt, g in prod.groupby("record_type"):
    names = sorted(g["provider_name"].unique())
    print(f"  {rt:<12} {len(names)} name(s), {len(g):>2} rows   {', '.join(names)[:70]}")
```

```python
# ============================================================================
# CELL 5: Reconciliation — verify the CSVs before anything reads them
# ============================================================================

def reconcile(pnl, tax, tol=0.02):
    """Check each CSV against its own internal arithmetic.

    Two independent identities, both of which must hold for every year:
      Net                          == Credit - Debit
      Total Discretionary Earnings == Net + the six P&L add-backs

    The second is the one that matters. That column is P&L-only by convention, so a year
    that fails by exactly the tax add-back amount means tax figures were folded into a
    P&L total -- which double-counts the moment anything sums the two files.

    tol is 2 cents rather than 1: FY2025 carries a known one-cent rounding difference
    inherited from the Decision Record, and a tolerance that flags it every run trains
    you to ignore the check.
    """
    rows, failures = [], []

    for year in pnl.index:
        r = pnl.loc[year]

        net_calc  = r["Credit"] - r["Debit"]
        net_delta = net_calc - r["Net"]

        addbacks   = r[PNL_ADDBACKS].sum()      # NaN skipped: a missing line is not an add-back
        total_calc = r["Net"] + addbacks
        tot_delta  = total_calc - r["Total Discretionary Earnings"]

        rows.append({"Year": year,
                     "Net": r["Net"],
                     "P&L add-backs": addbacks,
                     "Total (file)": r["Total Discretionary Earnings"],
                     "Total (calc)": total_calc,
                     "delta": tot_delta})

        if abs(net_delta) > tol:
            failures.append(f"  {year} Net != Credit - Debit by {net_delta:,.2f}")
        if abs(tot_delta) > tol:
            hint = ""
            if year in tax.index and abs(abs(tot_delta) - tax.loc[year, "Tax_Addbacks"]) < tol:
                hint = f"  <- equals the {year} tax add-back; tax figures folded into a P&L total"
            failures.append(f"  {year} Total != Net + add-backs by {tot_delta:,.2f}{hint}")

    report = pd.DataFrame(rows).set_index("Year")
    print(report.to_string(float_format=lambda v: f"{v:,.2f}"))

    if failures:
        print("\nRECONCILIATION FAILED")
        print("\n".join(failures))
        raise ValueError("CSV does not reconcile — fix the file before proceeding")

    print("\nReconciles to the cent, all years.")
    return report


def normalized_earnings(pnl, tax):
    """Net income + P&L add-backs + tax add-backs, per method §3.

    Tax add-backs come from the return, not the P&L, so a year without a return carries
    none -- FY2025 is genuinely P&L-only, not missing data.
    """
    out = pd.DataFrame(index=pnl.index)
    out["Net income"]    = pnl["Net"]
    out["P&L add-backs"] = pnl[PNL_ADDBACKS].sum(axis=1)
    out["Tax add-backs"] = tax["Tax_Addbacks"].reindex(pnl.index).fillna(0.0)
    out["Normalized earnings"] = out.sum(axis=1)
    return out


reconcile(pnl, tax)

print()
norm = normalized_earnings(pnl, tax)
print(norm.to_string(float_format=lambda v: f"{v:,.2f}"))
print(f"\nFour-year total: {norm['Normalized earnings'].sum():,.2f}")
```

```python
# ============================================================================
# CELL 6: Provider metrics — deterministic, per-provider
# ============================================================================
import re


def audit_productivity(df=None):
    """Flag rows where production / patients_seen disagrees with the stated avg_visit_fee.

    The three columns are redundant by construction, so a disagreement means one of them
    is wrong. Which one is not determinable from the file -- report, do not repair.
    """
    d = (prod if df is None else df).copy()
    d["calc_fee"] = (d["production"] / d["patients_seen"]).where(d["patients_seen"] > 0)
    bad = d[(d["calc_fee"] - d["avg_visit_fee"]).abs() > 0.02]
    return bad[["year", "provider_name", "patients_seen", "production",
                "avg_visit_fee", "calc_fee"]]


def resolve_provider(query, df=None):
    """Map a name fragment to an exact provider_name.

    Returns (name, note). name is None when nothing matches or the match is ambiguous;
    note says which. Credentials vary in the export ('DMI', 'RI'), so matching is on
    name tokens rather than the full string.
    """
    df = prod if df is None else df
    names = sorted(df["provider_name"].unique())
    q = query.lower().strip()

    exact = [n for n in names if n.lower() == q]
    if exact:
        return exact[0], ""

    contained = [n for n in names if n.lower() in q or q in n.lower()]
    if len(contained) == 1:
        return contained[0], ""

    tokens = [t for t in re.findall(r"[a-z]+", q) if len(t) > 2]
    scored = []
    for n in names:
        parts = set(re.findall(r"[a-z]+", n.lower()))
        hits = sum(1 for t in tokens if t in parts)
        if hits:
            scored.append((hits, n))
    if not scored:
        return None, f"no provider matching {query!r}"

    best = max(h for h, _ in scored)
    winners = [n for h, n in scored if h == best]
    if len(winners) > 1:
        return None, f"ambiguous: {', '.join(winners)}"
    return winners[0], ""


def provider_facts(name):
    """Per-year facts for one provider, plus the caveats that apply to reading them."""
    resolved, note = resolve_provider(name)
    if resolved is None:
        return {"provider": None, "error": note}

    g = prod[prod["provider_name"] == resolved].set_index("year").sort_index()
    rec = g["record_type"].iloc[0]

    table = g[["production", "collection", "patients_seen", "adjustments",
               "avg_visit_fee"]].copy()
    # Recomputed alongside the stated figure rather than replacing it: where they differ,
    # the row is unreliable and the caveat says so.
    table["fee_check"] = (table["production"] / table["patients_seen"]).where(
        table["patients_seen"] > 0)

    caveats = []
    suspect = (table["fee_check"] - table["avg_visit_fee"]).abs() > 0.02
    if suspect.any():
        yrs = ", ".join(str(y) for y in table.index[suspect])
        caveats.append(
            f"{yrs}: production / patients_seen disagrees with the stated avg_visit_fee. "
            "One of the three columns is wrong for that year; totals and averages "
            "spanning it are unreliable.")

    if rec != "individual":
        caveats.append(
            f"'{resolved}' is a {rec} record, not an individual provider (method §5.5). "
            "It must not be counted in per-provider averages or headcounts.")

    tot_prod, tot_pat = table["production"].sum(), table["patients_seen"].sum()

    # Collection is attributed on a different basis than production in this export, so a
    # per-provider rate is not computable. Method §5.5.
    if tot_prod > 0:
        ratio = table["collection"].sum() / tot_prod
        if ratio < 0.05 or ratio > 1.2:
            caveats.append(
                "Collection is not attributed to this provider on the same basis as "
                "production, so a per-provider collection rate is not computable. "
                "Use the practice-level rate.")

    return {
        "provider": resolved,
        "role": g["service_role"].iloc[0],
        "record_type": rec,
        "years": list(table.index),
        "table": table,
        "total_production": tot_prod,
        "total_patients": int(tot_pat),
        # Weighted, not an unweighted mean of yearly rates: that would weight a 7-visit
        # year the same as a 2,239-visit year.
        "avg_visit_fee_overall": (tot_prod / tot_pat) if tot_pat else None,
        "reliable": not suspect.any() and rec == "individual",
        "caveats": caveats,
    }


def provider_list(individuals_only=True):
    """Named clinicians, or every record in the export."""
    df = prod[prod["record_type"] == "individual"] if individuals_only else prod
    return sorted(df["provider_name"].unique())

def provider_fact_lines():
    """One line per provider, name adjacent to every figure.

    A wide pivot asks the model to hold a column heading in mind while reading four rows
    beneath it, which is where the misattribution came from — right figures, wrong
    column. A line per provider makes the association local and unmissable.
    """
    share = production_share("Dentist")
    doll  = prod[prod.service_role == "Dentist"].pivot_table(
        index="year", columns="provider_name", values="production", aggfunc="sum").fillna(0)

    owners = {"Harvey Seybold, DDS", "Peter S. Chang, DDS"}
    lines = []
    for name in share.columns:
        rt   = prod.loc[prod.provider_name == name, "record_type"].iloc[0]
        role = ("owner dentist" if name in owners else
                "non-owner dentist" if rt == "individual"
                else f"{rt} record, not an individual")
        pct = "; ".join(f"{y} {share.loc[y, name]:.1f}%" for y in share.index)
        amt = "; ".join(f"{y} {doll.loc[y, name]:,.2f}" for y in doll.index)
        lines.append(f"{name} — {role}. Share of dentist production: {pct}. "
                     f"Production in dollars: {amt}.")
    return lines


def _reps(v):
    """Every rounding of a figure a reader might see, so a quoted 44.5 matches 44.48."""
    v = float(v)
    return {round(v, 2), round(v, 1), round(v, 0)}


def attribution_errors(text, tol=0.005):
    """Flag figures stated next to the wrong provider's name.

    check() cannot catch this class: the numbers are all present in the source, just
    bound to the wrong person.

    Deliberately narrow. A figure is tested only when exactly one provider is named in
    its sentence — in "Chang drew X against Seybold's Y and produced 1.9%", nothing local
    says which name the 1.9% attaches to, and guessing produces false failures on correct
    prose. Missing those is better than crying wolf on them.
    """
    share = production_share("Dentist")
    doll  = prod[prod.service_role == "Dentist"].pivot_table(
        index="year", columns="provider_name", values="production", aggfunc="sum").fillna(0)

    owned = {}
    for n in share.columns:
        s = set()
        for v in list(share[n]) + list(doll[n]):
            s |= _reps(v)
        owned[n] = s

    surnames = {n: n.split(",")[0].split()[-1] for n in share.columns}
    problems = []

    for sentence in re.split(r'(?<=[.;])\s+', text):
        named = [n for n, sur in surnames.items()
                 if re.search(r'\b' + re.escape(sur) + r'\b', sentence)]
        if len(named) != 1:            # zero or ambiguous — not decidable locally
            continue
        name = named[0]
        for tok in re.findall(r'\d[\d,]*(?:\.\d+)?', sentence):
            try:
                val = round(float(tok.replace(",", "")), 2)
            except ValueError:
                continue
            if 1900 < val < 2100:
                continue
            if any(abs(val - o) < tol for o in owned[name]):
                continue
            others = [n for n, s in owned.items()
                      if n != name and any(abs(val - o) < tol for o in s)]
            if others:
                problems.append(f"{tok} attributed to {name}; belongs to {', '.join(others)}")
    return problems


issues = audit_productivity()
print(f"productivity audit: {len(issues)} inconsistent row(s) of {len(prod)}")
if len(issues):
    print(issues.to_string(index=False, float_format=lambda v: f"{v:,.2f}"))

print(f"\nindividuals: {len(provider_list())} | all records: {len(provider_list(False))}")
for q in ["Emily Hawkins, RDH", "Seybold", "Chang", "Bob Smith"]:
    n, note = resolve_provider(q)
    print(f"  {q:<20} -> {n or '(none)'} {note}")
```

```python
# ============================================================================
# CELL 7: Practice metrics — financial and production, practice-wide
# ============================================================================

PARTIAL_YEARS = {2025: "~149 days"}      # method §5.4: not annualized, not comparable

MEASURE = "normalized owner earnings"
MEASURE_NOTE = (
    "Owner compensation is normalized to a market rate and only the excess is added back, "
    "so this is normalized owner earnings, not SDE as the term is used in practice sales "
    "(method §1). The delivered files are labeled SDE.")


def earnings_by_year():
    """Normalized owner earnings per year, with its components. Method §3."""
    out = normalized_earnings(pnl, tax)
    out["partial year"] = [PARTIAL_YEARS.get(y, "") for y in out.index]
    return out


def addback_components(year=None):
    """The six P&L add-backs plus the two tax add-backs, by year or across all years.

    Kept separate by source: they come from different documents, and only the P&L file
    is competent for the former (method §2).
    """
    cols = pnl[PNL_ADDBACKS]
    out = (cols if year is None else cols.loc[[year]]).copy()
    out["Depreciation"] = tax["Depreciation"].reindex(out.index)
    out["Section 179"]  = tax["Section179"].reindex(out.index)
    return out


def practice_totals():
    """Production, collection and visits by year, over every record in the export.

    Practice-level only. Collection is attributed on a different basis than production
    per provider, so the rate is meaningful here and not per clinician (method §5.5).
    """
    g = prod.groupby("year").agg(production=("production", "sum"),
                                 collection=("collection", "sum"),
                                 visits=("patients_seen", "sum"))
    g["collection rate %"] = g["collection"] / g["production"] * 100
    g["visits"] = g["visits"].astype(int)
    g["partial year"] = [PARTIAL_YEARS.get(y, "") for y in g.index]
    return g


def production_share(role="Dentist"):
    """Share of production within one service role, by provider and year.

    Includes non-individual records where they carry production: the contractor line is
    real production, and dropping it would overstate every other share. It is labeled,
    not hidden. Note a 0.0 means a row with no production; an absent provider means no
    row at all.
    """
    d = prod[prod["service_role"] == role]
    piv = d.pivot_table(index="year", columns="provider_name",
                        values="production", aggfunc="sum").fillna(0)
    return piv.div(piv.sum(axis=1), axis=0) * 100


def valuation_range(*_, **__):
    """Deliberately not computed. Three reasons, all structural rather than temporary."""
    e = earnings_by_year()["Normalized earnings"]
    return {
        "computed": False,
        "reasons": [
            MEASURE_NOTE + " A multiple applied to this measure does not answer what a "
            "buyer's multiple answers; the figure must be recomputed under a full-add-back "
            "convention first.",
            "Full add-back requires the owner's total compensation, which is not in these "
            "files — only the excess over benchmark is. The converted figures are in the "
            "engagement record.",
            f"{int((e < 0).sum())} of {len(e)} years are negative; a multiple applied to a "
            "negative figure yields a negative valuation, which is not a result.",
        ],
        "earnings_by_year": e.to_dict(),
    }


print("Normalized owner earnings")
print(earnings_by_year().to_string(float_format=lambda v: f"{v:,.2f}"))
print("\nPractice totals")
print(practice_totals().to_string(float_format=lambda v: f"{v:,.2f}"))
print("\nDentist production share (%)")
print(production_share("Dentist").to_string(float_format=lambda v: f"{v:.1f}"))
```

```python
# ============================================================================
# CELL 8: Report sections — decide what the data supports, before any prose
# ============================================================================

def _fmt(df, dp=2):
    return df.to_string(float_format=lambda v: f"{v:,.{dp}f}")


def sec_measure():
    """Which measure was computed, and what it is not. Method §1."""
    e = earnings_by_year()["Normalized earnings"]
    return {"figures": _fmt(e.to_frame()),
            "notes": [MEASURE_NOTE],
            "query": "choosing the measure normalized owner earnings versus SDE labeling"}


def sec_financials():
    """Per-year lines rather than a pivot.

    The model reads a five-column grid by tracking position and loses its place — it has
    twice reported FY2023's collection rate as FY2024's. Labeling every value with its
    year removes the positional read.
    """
    t = practice_totals()
    lines = []
    for y in t.index:
        r = t.loc[y]
        partial = f" (partial year, {PARTIAL_YEARS[y]})" if y in PARTIAL_YEARS else ""
        lines.append(
            f"FY{y}{partial}: production {r['production']:,.2f}; "
            f"collection {r['collection']:,.2f}; visits {int(r['visits']):,}; "
            f"collection rate {r['collection rate %']:.2f}%.")

    return {"figures": "\n".join(lines),
            "notes": [f"FY{y} is a partial year ({d}); figures are not annualized and "
                      f"not comparable to full years (method §5.4)."
                      for y, d in PARTIAL_YEARS.items()],
            "query": "production collection partial year proration"}


def sec_earnings():
    return {"figures": _fmt(earnings_by_year()),
            "notes": ["Tax-side add-backs come from the return, so a year without a "
                      "return carries none (method §2)."],
            "query": "formula net income add-backs depreciation section 179"}


def sec_addbacks():
    comp = addback_components()
    zero = [c for c in comp.columns if comp[c].fillna(0).eq(0).all()]

    # NaN means "no line in the source", which is not zero. Rendered as a dash so the
    # prompt cannot read it as an amount, with the reason stated rather than implied.
    shown = comp.copy()
    figures = shown.to_string(float_format=lambda v: f"{v:,.2f}", na_rep="—")

    notes = ["A dash means the category has no line in the source document for that "
             "year, which is not the same as an amount of zero."]
    if any(comp[c].isna().any() for c in ("Depreciation", "Section 179")):
        notes.append("FY2025 has no depreciation or Section 179 because no FY2025 tax "
                     "return was available (method §2).")
    if zero:
        notes.append("Categories with no amount in any year: " + ", ".join(zero))

    return {"figures": figures, "notes": notes,
            "query": "add-back categories accepted rejected estimated ledger source documents tax return"}


def sec_providers():
    share  = production_share("Dentist")
    issues = audit_productivity()
    notes  = []
    nonind = sorted(prod.loc[prod.record_type != "individual", "provider_name"].unique())
    if nonind:
        notes.append(f"Non-individual records present: {', '.join(nonind)}. These are "
                     "record types, not clinicians, and are excluded from headcounts and "
                     "per-provider averages (method §5.5).")
    for _, r in issues.iterrows():
        notes.append(f"{r.provider_name} FY{int(r.year)}: production / patients_seen "
                     f"({r.production / r.patients_seen:,.2f}) disagrees with the stated "
                     f"avg_visit_fee ({r.avg_visit_fee:,.2f}). Figures spanning that year "
                     "are unreliable.")
    notes.append("Collection is attributed on a different basis than production per "
                 "provider, so collection rate is reported practice-wide only.")
    return {"figures": "\n".join(provider_fact_lines()),
            "notes": notes,
            "query": "provider record types non-owner clinician concentration"}


def sec_valuation():
    v = valuation_range()
    return {"figures": "No multiple applied.",
            "notes": v["reasons"],
            "query": "measure labeling relabel recompute before presenting to a buyer"}


def sec_open_items():
    """Corpus-carried, with the production figures the rules apply to.

    States what is present and what is absent. Whether an absence is consequential is
    what the retrieved rules decide -- asserting it here would leave the model
    paraphrasing this note and citing a rule it never needed.
    """
    d = prod[(prod.service_role == "Dentist") & (prod.record_type == "individual")]
    piv = d.pivot_table(index="year", columns="provider_name",
                        values="production", aggfunc="sum").fillna(0)

    owners = ["Harvey Seybold, DDS", "Peter S. Chang, DDS"]
    non_owner = [c for c in piv.columns if c not in owners]

    return {
        "figures": "Dentist production, dollars\n" + _fmt(piv),
        "notes": [
            f"Owners per the engagement record: {', '.join(owners)}. "
            f"Non-owner clinicians in this table: {', '.join(non_owner)}.",
            "No supplied file contains compensation for any non-owner clinician. "
            "The P&L carries owner wages only; the productivity export carries no "
            "dollar figure that enters the calculation.",
            "Health insurance was excluded from the calculation as an unsubstantiated "
            "estimate rather than substantiated from the ledger.",
        ],
        "query": "open items non-owner clinician concentration unsubstantiated estimate",
    }


SECTION_SPEC = [
    ("measure",    "Measure and labeling",        sec_measure),
    ("financials", "Key financial indicators",    sec_financials),
    ("earnings",   "Normalized earnings by year", sec_earnings),
    ("addbacks",   "Add-back detail",             sec_addbacks),
    ("providers",  "Provider production",         sec_providers),
    ("valuation",  "Valuation multiples",         sec_valuation),
    ("open_items", "Open items",                  sec_open_items),
]


def build_sections():
    """Run every section builder. A section survives if it has figures or notes.

    Deciding this in code rather than in the prompt is the point: a template that asks
    the model to fill a heading gets a filled heading whether or not the data supports one.
    """
    out = []
    for key, title, fn in SECTION_SPEC:
        try:
            s = fn()
        except Exception as e:
            print(f"  SKIP {key}: {type(e).__name__}: {e}")
            continue
        s.update(key=key, title=title)
        s["supported"] = bool(s["figures"].strip() or s["notes"])
        out.append(s)
    return out


sections = build_sections()
for s in sections:
    mark = "" if s["supported"] else "  (no data — carried by the corpus alone)"
    print(f"  {s['key']:<11} {s['title']:<28} {len(s['notes'])} note(s){mark}")
```

```python
# ============================================================================
# CELL 9: Guardrails
# ============================================================================

def report_check(rep):
    """One-line verdict plus whatever failed. Display only — the checks live in rag."""
    print(f"GUARDRAIL : {'PASS' if rep['pass'] else 'FAIL'}")
    for key in ("fabricated_citations", "malformed_citations", "misattributed_figures",
                "unsupported_numbers", "leaked_in_refusal", "uncited"):
        if rep.get(key):
            print(f"    {key}: {rep[key]}")


# Confirm the checks fire rather than passing everything by default.
_bad = rag.check("The FY2022 excess was 41,200.00 [roselle-99].", ["roselle-06"])
assert not _bad["pass"] and _bad["fabricated_citations"] == ["roselle-99"]
print("guardrails live |", rag.summary())
```

```python
# ============================================================================
# CELL 10: Section generation — retrieve, prompt, check
# ============================================================================

SECTION_SYSTEM = """You write one section of a dental practice valuation report.

You are given FIGURES (computed from the practice's source files), NOTES (facts about
those figures and about what the files do not contain), and SOURCES (numbered excerpts
from the firm's methodology and this engagement's record).

FIGURES and NOTES are facts. SOURCES are the rules and findings that explain them.

Rules:
1. Quote every figure exactly as it appears in FIGURES or NOTES. Never round, restate to
   fewer decimal places, recalculate, or introduce a number that is not given.
2. Cite every methodological claim inline as [source-id]. Any statement about why a
   treatment is correct, what a rule requires, or what a finding means must carry a
   citation to the source it rests on.
3. If SOURCES do not support a claim, do not make it. Do not fill gaps from general
   knowledge about valuation or accounting.
4. Draw conclusions from the rules, do not repeat the notes. A note states a fact; your
   job is to say what the cited rule makes of it. Never state the same fact twice.
5. Do not add recommendations, next steps, or follow-up suggestions.
6. Write plain professional prose, two to four short paragraphs. Do not write a heading.
7. The FIGURES table is printed in the report directly above your text. Do not reproduce
   it, in prose or as a table, and do not walk through it year by year. Name a figure
   only where a cited rule bears on it — at most three or four in the whole section.
8. Cite only in the form [source-id], using an id exactly as it appears in brackets at
   the head of a SOURCE — for example [method-04]. The excerpts contain provenance marks
   and cross-references such as [doc · DR §4] or "method §2"; these are part of the
   source text, not citation ids, and must never be used as citations. Facts from FIGURES
   and NOTES need no citation at all — state them plainly, with no bracket of any kind.
   Never write [FIGURES] or [NOTES].
9. Output the finished section only. No analysis, no plan, no commentary about the task.
   The first words of your reply are the first words of the report."""

ARITHMETIC_CLAUSE = """
7. You may compute simple derived figures (differences, percentages, totals) from the
   figures given. State the inputs whenever you do."""

_UNCITABLE = re.compile(r'\s*\[(?:NOTES|FIGURES)\]')


def clean_section(text):
    """Remove [NOTES] and [FIGURES] labels the model attaches to data-derived facts.

    These are the only two uncitable sources in the prompt, and the model reliably labels
    facts drawn from them despite three prompt revisions telling it not to. Stripping is
    safe because the labels carry no information — the sentence is a plain statement of
    data either way. A real citation to a chunk id is never touched.
    """
    return _UNCITABLE.sub("", text)


def section_prompt(sec, hits):
    """Assemble one section's prompt. The chunk breadcrumb is dropped from the body
    because it is already the source label, and repeating it wastes context."""
    sources = "\n\n".join(
        f"[{c['id']}] {c['heading']}\n{c['text'].split(chr(10) * 2, 1)[-1]}"
        for _, c in hits)
    notes = "\n".join(f"- {n}" for n in sec["notes"]) or "(none)"
    return (f"SECTION: {sec['title']}\n\n"
            f"FIGURES:\n{sec['figures'].strip() or '(none)'}\n\n"
            f"NOTES:\n{notes}\n\n"
            f"SOURCES:\n\n{sources}\n\n"
            f"Write the body of this section.")


def generate_section(sec, k=4, max_tokens=1600):
    """Retrieve for this section, generate its prose, check the result."""
    hits   = rag.search(sec["query"], k=k)
    system = SECTION_SYSTEM + (ARITHMETIC_CLAUSE if ALLOW_MODEL_ARITHMETIC else "")
    text   = clean_section(call_llm(system, section_prompt(sec, hits),
                                    max_tokens=max_tokens))

    # Figures and notes join the retrieved chunks as supporting context: without them
    # every correctly-quoted CSV figure would flag as unsupported.
    context = sec["figures"] + "\n" + "\n".join(sec["notes"])
    rep = rag.check(text, [c["id"] for _, c in hits], extra_context=context)

    # Attribution is checked here rather than in rag.check(): it needs the provider data,
    # which the corpus does not have.
    if sec["key"] in ("providers", "open_items"):
        misattributed = attribution_errors(text)
        if misattributed:
            rep["misattributed_figures"] = misattributed
            rep["pass"] = False

    return {"key": sec["key"], "title": sec["title"], "text": text,
            "retrieved": [(round(s, 3), c["id"]) for s, c in hits], "check": rep}


def show_section(out):
    assert out is not None, "generate_section returned None — check its return statement"
    print(f"{'=' * 78}\n## {out['title']}\n")
    print(f"RETRIEVED : {out['retrieved']}")
    report_check(out["check"])
    print()
    print(rag.render(out["text"]))
```

```python
outputs = []
for sec in sections:
    out = generate_section(sec)
    outputs.append(out)
    show_section(out)
    time.sleep(3)

fails = [o["key"] for o in outputs if not o["check"]["pass"]]
print(f"\n{len(outputs) - len(fails)}/{len(outputs)} sections passed"
      + (f" | failed: {fails}" if fails else ""))
```

```python
# ============================================================================
# CELL 11: Report assembly — tables beside the prose
# ============================================================================

def df_to_markdown(df, dp=2, index_name=None):
    """Render a DataFrame as a markdown table.

    pandas to_string() is fixed-width text: fine in a terminal, unreadable once the
    report is markdown. Numerics are right-aligned and missing values render as an em
    dash, matching the convention the prose refers to.
    """
    d    = df.copy()
    idx  = index_name or (d.index.name or "")
    head = [str(idx)] + [str(c) for c in d.columns]

    rows = []
    for i, r in d.iterrows():
        cells = []
        for v in r:
            if isinstance(v, float) and pd.isna(v):
                cells.append("—")
            # Integers before floats: bool and int both satisfy the float branch's intent,
            # and a visit count rendered as 6,787.00 reads as a currency amount.
            elif isinstance(v, (int, np.integer)) and not isinstance(v, bool):
                cells.append(f"{int(v):,}")
            elif isinstance(v, (float, np.floating)):
                cells.append(f"{v:,.{dp}f}")
            else:
                cells.append(str(v) if str(v).strip() else "")
        rows.append([str(i)] + cells)

    align = ["| :--- "] + [" ---: " for _ in d.columns]
    return "\n".join(["| " + " | ".join(head) + " |", "|".join(align) + "|"]
                     + ["| " + " | ".join(r) + " |" for r in rows])


def dentist_dollars():
    d = prod[(prod.service_role == "Dentist") & (prod.record_type == "individual")]
    return d.pivot_table(index="year", columns="provider_name",
                         values="production", aggfunc="sum").fillna(0)


# Which table belongs beside which section's prose. valuation maps to None — the absence
# of a figure is the finding there.
FRAMES = {
    "measure":    (lambda: earnings_by_year()[["Normalized earnings"]],     2),
    "financials": (practice_totals,                                          2),
    "earnings":   (lambda: earnings_by_year().drop(columns="partial year"),  2),
    "addbacks":   (addback_components,                                       2),
    "providers":  (lambda: production_share("Dentist"),                      1),
    "valuation":  (None,                                                     0),
    "open_items": (dentist_dollars,                                          2),
}

TABLE_CAPTION = {
    "measure":    "Normalized owner earnings by fiscal year",
    "financials": "Production, collections and visits by fiscal year",
    "earnings":   "Components of normalized owner earnings",
    "addbacks":   "Add-backs by category and fiscal year",
    "providers":  "Dentist production share (%)",
    "open_items": "Dentist production, dollars (named individuals only)",
}


def render_report(outputs, title="Valuation Analysis", subtitle=""):
    """Interleave heading, table, and prose for every section.

    The prompt tells the model the table is printed above its text. Until this function
    existed that was false, and the prose compensated by reciting figures the reader
    could not see.
    """
    parts = [f"# {title}"]
    if subtitle:
        parts.append(f"*{subtitle}*")

    for out in outputs:
        parts.append(f"\n## {out['title']}")
        fn, dp = FRAMES.get(out["key"], (None, 2))
        if fn is not None:
            if out["key"] in TABLE_CAPTION:
                parts.append(f"**{TABLE_CAPTION[out['key']]}**\n")
            parts.append(df_to_markdown(fn(), dp=dp))
            parts.append("")
        parts.append((rag.render(out["text"]).strip()))

    return "\n".join(parts)

from IPython.display import Markdown, display

report_md = render_report(
    outputs,
    title="Roselle Dental Center, P.C. — Normalized Earnings Analysis",
    subtitle="FY2022–FY2025 · prepared for the owners")  # clean, real $ signs

# The file is correct as-is.
Path("Roselle_Report.md").write_text(report_md, encoding="utf-8")


def display_report(md):
    """Show the report in Jupyter without MathJax eating currency.

    Two defenses because the class name is version-dependent: MathJax 2 honors
    tex2jax_ignore, MathJax 3 (JupyterLab 3+, Notebook 7) honors mathjax_ignore. Wrapping
    each $ in its own element breaks delimiter pairing regardless, since MathJax needs
    both delimiters in one text node.
    """
    from IPython.display import HTML, display
    import markdown as md_lib
    html = md_lib.markdown(md, extensions=["tables"])
    html = html.replace("$", "<span>$</span>")
    display(HTML(f'<div class="mathjax_ignore tex2jax_ignore">{html}</div>'))


display_report(report_md)
```

```python
# ============================================================================
# CELL 3: Enhanced Data Loading and Validation Functions
# ============================================================================

class DentalPracticeData:
    """Enhanced class to handle dental practice data from multiple sources"""
    
    def __init__(self, 
                 productivity_csv: Optional[str] = None,
                 taxes_sde_csv: Optional[str] = None,
                 pnl_sde_csv: Optional[str] = None,
                 procedures_csv: Optional[str] = None,
                 filter_years: Optional[List[int]] = None):
        """
        Initialize with CSV files for productivity and SDE calculations
        
        Args:
            productivity_csv: Path to productivity CSV file
            taxes_sde_csv: Path to tax-based SDE calculation CSV (2022-2024)
            pnl_sde_csv: Path to P&L-based SDE calculation CSV (2022-2025)
            procedures_csv: Path to procedures CSV file (optional, on hold)
            filter_years: List of years to include in analysis (None = all years)  # ADD THIS
        """
        # ADD THIS LINE:
        self.filter_years = filter_years
        
        # Initialize data storage
        self.data = {}
        
        # Load productivity data if provided
        self.productivity_data = None
        if productivity_csv:
            self.productivity_data = self._load_productivity_data(productivity_csv)
            
        # Load SDE data from taxes if provided
        self.taxes_sde_data = None
        if taxes_sde_csv:
            self.taxes_sde_data = self._load_taxes_sde_data(taxes_sde_csv)
            
        # Load SDE data from P&L if provided
        self.pnl_sde_data = None
        if pnl_sde_csv:
            self.pnl_sde_data = self._load_pnl_sde_data(pnl_sde_csv)
            
        # Load procedures data if provided (on hold for now)
        self.procedures_data = None
        if procedures_csv:
            self.procedures_data = self._load_procedures_data(procedures_csv)
            
        # ADD THIS LINE - Apply year filtering if specified:
        self._apply_year_filter()
        
        # Calculate derived metrics and prepare valuation data
        self._calculate_sde_metrics()
        self._calculate_derived_metrics()
        self._prepare_valuation_data()
    
    def _apply_year_filter(self):
        """Apply year filtering to all loaded datasets"""
        if not self.filter_years:
            return
            
        # Filter productivity data
        if self.productivity_data is not None and 'year' in self.productivity_data.columns:
            self.productivity_data = self.productivity_data[
                self.productivity_data['year'].isin(self.filter_years)
            ]
            
        # Filter taxes SDE data
        if self.taxes_sde_data is not None:
            year_col = 'Year' if 'Year' in self.taxes_sde_data.columns else 'year'
            if year_col in self.taxes_sde_data.columns:
                self.taxes_sde_data = self.taxes_sde_data[
                    self.taxes_sde_data[year_col].isin(self.filter_years)
                ]
                
        # Filter P&L SDE data
        if self.pnl_sde_data is not None:
            year_col = 'Year' if 'Year' in self.pnl_sde_data.columns else 'year'
            if year_col in self.pnl_sde_data.columns:
                self.pnl_sde_data = self.pnl_sde_data[
                    self.pnl_sde_data[year_col].isin(self.filter_years)
                ]
                
        # Filter procedures data if present
        if self.procedures_data is not None and 'year' in self.procedures_data.columns:
            self.procedures_data = self.procedures_data[
                self.procedures_data['year'].isin(self.filter_years)
            ]
    
    def get_available_years(self) -> List[int]:
        """Get list of all years available in the datasets"""
        years = set()
        
        if self.productivity_data is not None and 'year' in self.productivity_data.columns:
            years.update(self.productivity_data['year'].unique())
            
        if self.taxes_sde_data is not None:
            year_col = 'Year' if 'Year' in self.taxes_sde_data.columns else 'year'
            if year_col in self.taxes_sde_data.columns:
                years.update(self.taxes_sde_data[year_col].unique())
                
        if self.pnl_sde_data is not None:
            year_col = 'Year' if 'Year' in self.pnl_sde_data.columns else 'year'
            if year_col in self.pnl_sde_data.columns:
                years.update(self.pnl_sde_data[year_col].unique())
                
        return sorted(list(years))
        
    def _load_productivity_data(self, csv_path: str) -> pd.DataFrame:
        """Load and process productivity CSV data"""
        try:
            df = pd.read_csv(csv_path)
            
            # Handle column name variations
            if 'provider_name' in df.columns and 'provider' not in df.columns:
                df = df.rename(columns={'provider_name': 'provider'})
            
            print(f"Productivity CSV columns: {list(df.columns)}")
            
            # Clean numeric columns
            numeric_cols = ['patients_seen', 'avg_visit_fee', 'production', 
                           'collection', 'adjustments']
            for col in numeric_cols:
                if col in df.columns:
                    df[col] = pd.to_numeric(df[col].astype(str).str.replace('$', '').str.replace(',', ''), errors='coerce')
            
            print(f"Loaded productivity data: {len(df)} records")
            return df
        except Exception as e:
            print(f"Error loading productivity data: {e}")
            return None
    
    def _load_taxes_sde_data(self, csv_path: str) -> pd.DataFrame:
        """Load and process tax-based SDE calculation data"""
        try:
            df = pd.read_csv(csv_path)
            print(f"Taxes SDE CSV columns: {list(df.columns)}")
            
            # Clean numeric columns - handle various formats
            for col in df.columns:
                if col not in ['Year', 'year']:
                    # Try to convert to numeric, handling currency symbols
                    df[col] = pd.to_numeric(
                        df[col].astype(str).str.replace('$', '').str.replace(',', '').str.replace('(', '-').str.replace(')', ''),
                        errors='coerce'
                    )
            
            print(f"Loaded taxes SDE data: {len(df)} records for years {df.get('Year', df.get('year', [])).unique()}")
            return df
        except Exception as e:
            print(f"Error loading taxes SDE data: {e}")
            return None
    
    def _load_pnl_sde_data(self, csv_path: str) -> pd.DataFrame:
        """Load and process P&L-based SDE calculation data"""
        try:
            df = pd.read_csv(csv_path)
            print(f"P&L SDE CSV columns: {list(df.columns)}")
            
            # First, clean up the Year column and remove any rows with invalid years
            if 'Year' in df.columns:
                df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
                df = df.dropna(subset=['Year'])  # Remove rows with NaN years
                df['Year'] = df['Year'].astype(int)
            elif 'year' in df.columns:
                df['year'] = pd.to_numeric(df['year'], errors='coerce')
                df = df.dropna(subset=['year'])  # Remove rows with NaN years
                df['year'] = df['year'].astype(int)
            
            # Clean numeric columns - handle NA values
            numeric_columns = ['Automobile', 'Contributions', 'Legal Fees', 'Miscellaneous',
                              'Travel & Entertainment', 'Owner Excess Compensation', 
                              'Debit', 'Credit', 'Net', 'Total Discretionary Earnings']
            
            for col in numeric_columns:
                if col in df.columns:
                    # Replace 'NA' string with 0, handle NaN
                    df[col] = df[col].replace(['NA', 'na', 'N/A', 'n/a'], 0)
                    df[col] = pd.to_numeric(
                        df[col].astype(str).str.replace('$', '').str.replace(',', '').str.replace('(', '-').str.replace(')', ''),
                        errors='coerce'
                    ).fillna(0)
            
            year_col = 'Year' if 'Year' in df.columns else 'year'
            print(f"Loaded P&L SDE data: {len(df)} records for years {sorted(df[year_col].unique())}")
            return df
        except Exception as e:
            print(f"Error loading P&L SDE data: {e}")
            return None
    
    def _load_procedures_data(self, csv_path: str) -> pd.DataFrame:
        """Load and process procedures CSV data (keeping existing logic)"""
        try:
            df = pd.read_csv(csv_path)
            print(f"Procedures CSV columns: {list(df.columns)}")
            
            # Clean service names if present
            if 'service' in df.columns:
                df['service_clean'] = df['service'].apply(self._clean_service_name)
                df['service_category'] = df['service_clean'].apply(self._categorize_service)
            
            # Clean numeric columns
            numeric_cols = ['number', 'amount']
            for col in numeric_cols:
                if col in df.columns:
                    df[col] = pd.to_numeric(df[col].astype(str).str.replace('$', '').str.replace(',', ''), errors='coerce')
            
            print(f"Loaded procedures data: {len(df)} records")
            return df
        except Exception as e:
            print(f"Error loading procedures data: {e}")
            return None
    
    def _clean_service_name(self, service: str) -> str:
        """Clean service names using regex patterns (keeping existing logic)"""
        if pd.isna(service):
            return "Unknown"
        
        service = str(service).strip()
        
        # Common patterns to clean
        patterns = [
            (r'^D\d{4}\s*-?\s*', ''),  # Remove procedure codes at start
            (r'\s+', ' '),  # Multiple spaces to single
            (r'[^\w\s-]', ''),  # Remove special characters except dash
            (r'^\s*-\s*', ''),  # Remove leading dashes
        ]
        
        for pattern, replacement in patterns:
            service = re.sub(pattern, replacement, service)
        
        return service if service else "Unknown"
    
    def _categorize_service(self, service: str) -> str:
        """Categorize services into major groups (keeping existing logic)"""
        service_lower = service.lower()
        
        categories = {
            'Diagnostic': ['exam', 'x-ray', 'bitewing', 'panoramic', 'consult'],
            'Preventive': ['prophy', 'clean', 'fluoride', 'sealant'],
            'Restorative': ['filling', 'crown', 'bridge', 'composite', 'amalgam'],
            'Endodontic': ['root canal', 'endo', 'pulp'],
            'Periodontal': ['perio', 'scaling', 'root planing', 'gum'],
            'Oral Surgery': ['extraction', 'surgery', 'implant'],
            'Prosthodontic': ['denture', 'partial', 'prosth'],
            'Orthodontic': ['ortho', 'braces', 'aligner'],
        }
        
        for category, keywords in categories.items():
            if any(keyword in service_lower for keyword in keywords):
                return category
        
        return 'Other'
    
    def _calculate_sde_metrics(self):
        """Calculate SDE metrics from available data - FIXED VERSION"""
        print("\n=== Calculating SDE Metrics ===")
        
        # Initialize components
        sde_components = {}
        
        # First, get P&L-based addbacks
        if self.pnl_sde_data is not None and not self.pnl_sde_data.empty:
            print(f"P&L Data shape: {self.pnl_sde_data.shape}")
            
            # IMPORTANT: The current code sums ALL years together
            # This might not be what we want for valuation
            
            # Option 1: Keep total across all years (current behavior)
            pnl_addbacks = {
                'Owner_Excess_Compensation': self.pnl_sde_data['Owner Excess Compensation'].sum() if 'Owner Excess Compensation' in self.pnl_sde_data.columns else 0,
                'Automobile': self.pnl_sde_data['Automobile'].sum() if 'Automobile' in self.pnl_sde_data.columns else 0,
                'Contributions': self.pnl_sde_data['Contributions'].sum() if 'Contributions' in self.pnl_sde_data.columns else 0,
                'Legal_Fees': self.pnl_sde_data['Legal Fees'].sum() if 'Legal Fees' in self.pnl_sde_data.columns else 0,
                'Miscellaneous': self.pnl_sde_data['Miscellaneous'].sum() if 'Miscellaneous' in self.pnl_sde_data.columns else 0,
                'Travel_Entertainment': self.pnl_sde_data['Travel & Entertainment'].sum() if 'Travel & Entertainment' in self.pnl_sde_data.columns else 0,
            }
            sde_components.update(pnl_addbacks)
            print(f"P&L Addbacks (All Years Total): {pnl_addbacks}")
            
            # Get Net Income
            self.data['Net_Income_Total'] = self.pnl_sde_data['Net'].sum() if 'Net' in self.pnl_sde_data.columns else 0
            print(f"Net Income Total (All Years): ${self.data['Net_Income_Total']:,.2f}")
            
            # Option 2: Calculate average annual SDE (more useful for valuation)
            year_col = 'Year' if 'Year' in self.pnl_sde_data.columns else 'year'
            years = self.pnl_sde_data[year_col].unique()
            num_years = len(years)
            
            if num_years > 0:
                # Calculate average annual components
                self.data['Average_Annual_SDE_Components'] = {
                    k: v / num_years for k, v in pnl_addbacks.items()
                }
                self.data['Average_Annual_Net_Income'] = self.data['Net_Income_Total'] / num_years
                
                print(f"\nNumber of years in P&L data: {num_years}")
                print(f"Average Annual Net Income: ${self.data['Average_Annual_Net_Income']:,.2f}")
        
        # Add tax-based depreciation/Section179
        if self.taxes_sde_data is not None and not self.taxes_sde_data.empty:
            print(f"\nTax Data shape: {self.taxes_sde_data.shape}")
            print(f"Tax Columns: {list(self.taxes_sde_data.columns)}")
            
            tax_addbacks = {}
            
            # Check for Depreciation column
            if 'Depreciation' in self.taxes_sde_data.columns:
                tax_addbacks['Depreciation'] = self.taxes_sde_data['Depreciation'].sum()
            
            # Check for Section179
            if 'Section179' in self.taxes_sde_data.columns:
                tax_addbacks['Section179'] = self.taxes_sde_data['Section179'].sum()
            
            sde_components.update(tax_addbacks)
            print(f"Tax Addbacks (All Years Total): {tax_addbacks}")
            
            # Calculate average annual tax components
            tax_year_col = 'Year' if 'Year' in self.taxes_sde_data.columns else 'year'
            tax_years = self.taxes_sde_data[tax_year_col].unique()
            num_tax_years = len(tax_years)
            
            if num_tax_years > 0 and 'Average_Annual_SDE_Components' in self.data:
                for k, v in tax_addbacks.items():
                    self.data['Average_Annual_SDE_Components'][k] = v / num_tax_years
        
        # Store components
        self.data['SDE_Components'] = sde_components
        
        # Calculate total SDE: Net Income + All Addbacks
        net_income = self.data.get('Net_Income_Total', 0)
        total_addbacks = sum(sde_components.values())
        self.data['Total_SDE'] = net_income + total_addbacks
        self.data['Total_Addbacks'] = total_addbacks
        
        # Calculate average annual SDE
        if 'Average_Annual_Net_Income' in self.data and 'Average_Annual_SDE_Components' in self.data:
            avg_annual_addbacks = sum(self.data['Average_Annual_SDE_Components'].values())
            self.data['Average_Annual_SDE'] = self.data['Average_Annual_Net_Income'] + avg_annual_addbacks
            
            print(f"\n=== Average Annual SDE ===")
            print(f"Average Annual Net Income: ${self.data['Average_Annual_Net_Income']:,.2f}")
            print(f"Average Annual Add-backs: ${avg_annual_addbacks:,.2f}")
            print(f"Average Annual SDE: ${self.data['Average_Annual_SDE']:,.2f}")
        
        print(f"\n=== Total SDE Calculation Summary ===")
        print(f"Total Net Income (All Years): ${net_income:,.2f}")
        print(f"Total Addbacks (All Years): ${total_addbacks:,.2f}")
        print(f"Total SDE (All Years): ${self.data['Total_SDE']:,.2f}")
        print("="*40)
    
    def _calculate_derived_metrics(self):
        """Calculate additional metrics from the data"""
        # Calculate Adjusted Net Income (which is effectively the revenue metric we care about)
        if 'Net_Income_Total' in self.data and 'Total_Addbacks' in self.data:
            # Adjusted Net Income = Net Income + Addbacks
            self.data['Adjusted_Net_Income'] = self.data['Net_Income_Total'] + self.data['Total_Addbacks']
            
            # For valuation purposes, we might want to use the average of recent years
            if 'SDE_Historical' in self.data and self.data['SDE_Historical']:
                recent_years = sorted(self.data['SDE_Historical'].keys())[-3:]  # Last 3 years
                if recent_years:
                    recent_sde_values = [self.data['SDE_Historical'][year] for year in recent_years]
                    self.data['Average_SDE_Recent'] = sum(recent_sde_values) / len(recent_sde_values)
        
        # If we have P&L data, get revenue/collections info
        if self.pnl_sde_data is not None and not self.pnl_sde_data.empty:
            # Get Debit/Credit for gross revenue understanding
            if 'Debit' in self.pnl_sde_data.columns:
                self.data['Total_Debit'] = self.pnl_sde_data['Debit'].sum()
            if 'Credit' in self.pnl_sde_data.columns:
                self.data['Total_Credit'] = self.pnl_sde_data['Credit'].sum()
        
        # Use productivity data for actual collections/production if available
        if self.productivity_data is not None and 'production' in self.productivity_data.columns:
            # IMPORTANT: Calculate TOTAL production across ALL years
            self.data['Production_Total_All_Years'] = self.productivity_data['production'].sum()
            
            # Also calculate by year for context
            if 'year' in self.productivity_data.columns:
                yearly_production = self.productivity_data.groupby('year')['production'].sum()
                self.data['Production_By_Year'] = yearly_production.to_dict()
                
                # Get latest year for TTM (but note if it's partial)
                latest_year = self.productivity_data['year'].max()
                latest_year_data = self.productivity_data[self.productivity_data['year'] == latest_year]
                self.data['Production_Latest_Year'] = latest_year_data['production'].sum()
                self.data['Latest_Year'] = latest_year
                
                # Calculate average annual production (excluding partial 2025)
                complete_years = [y for y in yearly_production.index if y < 2025]
                if complete_years:
                    complete_years_production = [yearly_production[y] for y in complete_years]
                    self.data['Production_Average_Annual'] = sum(complete_years_production) / len(complete_years_production)
            
            # Collections data
            if 'collection' in self.productivity_data.columns:
                self.data['Collections_Total_All_Years'] = self.productivity_data['collection'].sum()
                
                if 'year' in self.productivity_data.columns:
                    yearly_collections = self.productivity_data.groupby('year')['collection'].sum()
                    self.data['Collections_By_Year'] = yearly_collections.to_dict()
                    
                    # Collection rate calculation
                    self.data['Collection_Rate_Overall'] = (
                        self.data['Collections_Total_All_Years'] / 
                        self.data['Production_Total_All_Years'] * 100
                    ) if self.data['Production_Total_All_Years'] > 0 else 0
                    
                    # Latest year collection rate
                    latest_year_collections = latest_year_data['collection'].sum() if 'collection' in latest_year_data.columns else 0
                    self.data['Collection_Rate_Latest_Year'] = (
                        latest_year_collections / 
                        self.data['Production_Latest_Year'] * 100
                    ) if self.data['Production_Latest_Year'] > 0 else 0
            
            # For TTM (Trailing Twelve Months), use the average annual production
            # This is more representative than using partial 2025 data
            self.data['Revenue_Trailing12'] = self.data.get('Production_Average_Annual', self.data.get('Production_Latest_Year', 0))
            self.data['Production_Trailing12'] = self.data['Revenue_Trailing12']
            
            # If we have collections data, use average annual collections
            if 'Collections_By_Year' in self.data and 'Production_Average_Annual' in self.data:
                complete_years = [y for y in self.data['Collections_By_Year'].keys() if y < 2025]
                if complete_years:
                    complete_collections = [self.data['Collections_By_Year'][y] for y in complete_years]
                    self.data['Collections_Trailing12'] = sum(complete_collections) / len(complete_collections)
                    self.data['Collection_Rate'] = (
                        self.data['Collections_Trailing12'] / 
                        self.data['Production_Average_Annual'] * 100
                    ) if self.data['Production_Average_Annual'] > 0 else 0
        
        # For SDE calculations, use the existing data
        if 'Total_SDE' in self.data:
            # Calculate TTM SDE based on average of complete years
            if self.pnl_sde_data is not None:
                year_col = 'Year' if 'Year' in self.pnl_sde_data.columns else 'year'
                years = self.pnl_sde_data[year_col].unique()
                complete_years = [y for y in years if y < 2025]
                
                if complete_years:
                    # This is already calculated in Total_SDE divided by number of years
                    num_years = len(complete_years)
                    self.data['SDE_Trailing12'] = self.data['Total_SDE'] / num_years if num_years > 0 else self.data['Total_SDE']
                else:
                    self.data['SDE_Trailing12'] = self.data['Total_SDE']
            
            # EBITDA approximation
            if 'SDE_Trailing12' in self.data:
                # EBITDA = SDE minus owner excess compensation
                owner_comp = self.data.get('SDE_Components', {}).get('Owner_Excess_Compensation', 0)
                # Divide by number of years to get annual
                num_years = len(self.data.get('Production_By_Year', {1: 1}))
                annual_owner_comp = owner_comp / num_years if num_years > 0 else owner_comp
                self.data['EBITDA_Trailing12'] = self.data['SDE_Trailing12'] - (annual_owner_comp * 0.5)  # Rough adjustment
        
        # Calculate margins if we have the data
        if 'SDE_Trailing12' in self.data and 'Production_Trailing12' in self.data:
            self.data['SDE_Margin'] = (self.data['SDE_Trailing12'] / 
                                      self.data['Production_Trailing12'] * 100) if self.data['Production_Trailing12'] > 0 else 0
        
        if 'EBITDA_Trailing12' in self.data and 'Production_Trailing12' in self.data:
            self.data['EBITDA_Margin'] = (self.data['EBITDA_Trailing12'] / 
                                         self.data['Production_Trailing12'] * 100) if self.data['Production_Trailing12'] > 0 else 0
    
    def _prepare_valuation_data(self):
        """Prepare data specifically for valuation report generation"""
        # Set practice information
        self.data['Practice_Overview'] = {
            'Name': 'Demo Dental Practice',  # This should be configurable
            'Location': 'Location TBD',
            'Valuation_Date': datetime.now().strftime('%Y-%m-%d'),
            'Operatories': 6,  # Default, should be configurable
        }
        
        # Calculate valuation multiples
        if 'Total_SDE' in self.data:
            # Industry standard multiples
            self.data['Valuation_Multiples'] = {
                'SDE_Multiple_Low': 3.5,
                'SDE_Multiple_High': 5.5,
                'SDE_Multiple_Mid': 4.5,
                'EBITDA_Multiple': 4.5,
                'Revenue_Multiple': 0.8
            }
            
            # Calculate implied valuations
            self.data['Implied_Valuations'] = {
                'SDE_Low': self.data['Total_SDE'] * self.data['Valuation_Multiples']['SDE_Multiple_Low'],
                'SDE_High': self.data['Total_SDE'] * self.data['Valuation_Multiples']['SDE_Multiple_High'],
                'SDE_Mid': self.data['Total_SDE'] * self.data['Valuation_Multiples']['SDE_Multiple_Mid'],
            }
            
            if 'EBITDA_Trailing12' in self.data:
                self.data['Implied_Valuations']['EBITDA'] = (
                    self.data['EBITDA_Trailing12'] * self.data['Valuation_Multiples']['EBITDA_Multiple']
                )
            
            if 'Revenue_Trailing12' in self.data:
                self.data['Implied_Valuations']['Revenue'] = (
                    self.data['Revenue_Trailing12'] * self.data['Valuation_Multiples']['Revenue_Multiple']
                )
    
    def get_comprehensive_context(self) -> str:
        """Get all available data as a formatted string for LLM context"""
        context = f"""
        === PRACTICE INFORMATION ===
        Name: {self.data.get('Practice_Overview', {}).get('Name', 'Unknown')}
        Location: {self.data.get('Practice_Overview', {}).get('Location', 'Unknown')}
        Valuation Date: {self.data.get('Practice_Overview', {}).get('Valuation_Date', 'Unknown')}
        
        === FINANCIAL METRICS ===
        Revenue (TTM - Average Annual): ${self.data.get('Revenue_Trailing12', 0):,.0f}
        Production (Total All Years): ${self.data.get('Production_Total_All_Years', 0):,.0f}
        Production (Average Annual): ${self.data.get('Production_Average_Annual', 0):,.0f}
        Collections (Total All Years): ${self.data.get('Collections_Total_All_Years', 0):,.0f}
        Collections (Average Annual): ${self.data.get('Collections_Trailing12', 0):,.0f}
        Collection Rate (Overall): {self.data.get('Collection_Rate_Overall', 0):.1f}%
        Collection Rate (Average): {self.data.get('Collection_Rate', 0):.1f}%
        EBITDA (TTM): ${self.data.get('EBITDA_Trailing12', 0):,.0f}
        Total SDE (All Years): ${self.data.get('Total_SDE', 0):,.0f}
        SDE (TTM - Annual Average): ${self.data.get('SDE_Trailing12', 0):,.0f}
        EBITDA Margin: {self.data.get('EBITDA_Margin', 0):.1f}%
        SDE Margin: {self.data.get('SDE_Margin', 0):.1f}%
        """
        
        # Add production by year
        if 'Production_By_Year' in self.data:
            context += "\n    === PRODUCTION BY YEAR ==="
            for year, prod in sorted(self.data['Production_By_Year'].items()):
                collections = self.data.get('Collections_By_Year', {}).get(year, 0)
                rate = (collections / prod * 100) if prod > 0 else 0
                context += f"\n    {year}: Production ${prod:,.0f}, Collections ${collections:,.0f} (Rate: {rate:.1f}%)"
                if year == 2025:
                    context += " [PARTIAL YEAR - through September]"
        
        # Add SDE components
        context += f"""
        
        === SDE COMPONENTS (Total All Years) ==="""
        
        if 'SDE_Components' in self.data:
            for component, value in self.data['SDE_Components'].items():
                context += f"\n    {component.replace('_', ' ')}: ${value:,.0f}"
        
        context += f"""
        
        === VALUATION ===
        SDE Multiple Range: {self.data.get('Valuation_Multiples', {}).get('SDE_Multiple_Low', 0)}-{self.data.get('Valuation_Multiples', {}).get('SDE_Multiple_High', 0)}x
        Implied Valuation (Low): ${self.data.get('Implied_Valuations', {}).get('SDE_Low', 0):,.0f}
        Implied Valuation (High): ${self.data.get('Implied_Valuations', {}).get('SDE_High', 0):,.0f}
        """
        
        # Add productivity context if available
        if self.productivity_data is not None and not self.productivity_data.empty:
            # Provider analysis by role
            if 'service_role' in self.productivity_data.columns:
                context += "\n\n    === PROVIDER ANALYSIS (All Years Combined) ==="
                role_summary = self.productivity_data.groupby('service_role').agg({
                    'provider': 'nunique',
                    'production': 'sum',
                    'patients_seen': 'sum'
                })
                
                for role, data in role_summary.iterrows():
                    avg_prod = data['production'] / data['provider'] if data['provider'] > 0 else 0
                    context += f"\n    {role}s: {int(data['provider'])} providers, ${data['production']:,.0f} total production"
                    context += f"\n        Average per provider (all years): ${avg_prod:,.0f}"
                    context += f"\n        Patients seen: {int(data['patients_seen']):,}"
                
                # Add year-by-year breakdown
                context += "\n\n    === PROVIDER PRODUCTION BY YEAR AND ROLE ==="
                for year in sorted(self.productivity_data['year'].unique()):
                    year_data = self.productivity_data[self.productivity_data['year'] == year]
                    context += f"\n    \n    {year}:"
                    year_summary = year_data.groupby('service_role').agg({
                        'provider': 'nunique',
                        'production': 'sum'
                    })
                    for role, data in year_summary.iterrows():
                        avg_prod = data['production'] / data['provider'] if data['provider'] > 0 else 0
                        context += f"\n        {role}s: {int(data['provider'])} providers, ${data['production']:,.0f} (avg: ${avg_prod:,.0f})"
            
            # Overall productivity metrics
            context += f"""
            
        === PRODUCTIVITY METRICS (All Years) ===
        Total Providers: {self.productivity_data['provider'].nunique() if 'provider' in self.productivity_data.columns else 0}
        Total Patients Seen: {self.productivity_data['patients_seen'].sum() if 'patients_seen' in self.productivity_data.columns else 0:,.0f}
        Average Visit Fee (Overall): ${self.productivity_data['avg_visit_fee'].mean() if 'avg_visit_fee' in self.productivity_data.columns else 0:.2f}
        """
        
        # Add historical SDE trend if available
        if 'SDE_Historical' in self.data and self.data['SDE_Historical']:
            context += "\n\n    === SDE TREND ==="
            for year, sde in sorted(self.data['SDE_Historical'].items()):
                context += f"\n    {year}: ${sde:,.0f}"
        
        return context
```

```python
# ============================================================================
# CELL 4: Enhanced LLM Interaction Functions with Valuation Report Generation
# ============================================================================

def categorize_question(question: str) -> str:
    """Enhanced question categorization with valuation categories"""
    question_lower = question.lower()
    
    # Extended category mappings
    category_keywords = {
        'valuation': ['valuation', 'value', 'worth', 'sde', 'ebitda', 'multiple', 
                     'price', 'asking', 'offer', 'appraisal', 'assessment'],
        'sde': ['sde', 'seller discretionary', 'add-back', 'adjustment', 
                'discretionary earnings', 'owner benefit'],
        'financial': ['revenue', 'profit', 'ebitda', 'margin', 'collection', 
                     'financial', 'p&l', 'income', 'expense'],
        'productivity': ['productivity', 'production', 'collection', 'efficiency', 
                        'patients seen', 'visit fee', 'utilization'],
        'procedures': ['procedure', 'service', 'treatment', 'bitewing', 'crown',
                      'filling', 'exam', 'frequency', 'procedure mix'],
        'report': ['report', 'summary', 'comprehensive', 'full analysis', 
                  'detailed', 'complete'],
        'staffing': ['hygienist', 'staff', 'overtime', 'employee', 'turnover',
                    'ratio', 'utilization', 'satisfaction'],
        'location': ['location', 'address', 'demographics', 'population', 
                    'competition', 'accessibility', 'area'],
        'tax': ['tax', 'depreciation', 'amortization', 'write'],
        'trend': ['trend', 'historical', 'growth', 'decline', 'change', 'year over year']
    }
    
    for category, keywords in category_keywords.items():
        if any(keyword in question_lower for keyword in keywords):
            return category
    
    return 'general'

def detect_provider_in_question(question: str, practice_data: DentalPracticeData) -> Optional[str]:
    """Detect if a specific provider is mentioned in the question"""
    if practice_data.productivity_data is None:
        return None
    
    # Get list of all providers from available data
    providers = set()
    if practice_data.productivity_data is not None and 'provider' in practice_data.productivity_data.columns:
        providers.update(practice_data.productivity_data['provider'].dropna().unique())
    
    question_lower = question.lower()
    best_match = None
    best_score = 0
    
    # Check each provider for matches
    for provider in providers:
        provider_lower = provider.lower()
        score = 0
        
        # Check for full name match
        if provider_lower in question_lower:
            return provider
        
        # Check for partial matches
        provider_parts = provider.split()
        for part in provider_parts:
            if len(part) > 2:
                import re
                pattern = r'\b' + re.escape(part.lower()) + r'\b'
                if re.search(pattern, question_lower):
                    score += 1
        
        if score > best_score:
            best_score = score
            best_match = provider
    
    # Return best match if we found at least 2 matching parts
    if best_score >= 2:
        return best_match
    
    return None

def generate_valuation_report_prompt(practice_data: DentalPracticeData) -> str:
    """Generate a prompt for creating a comprehensive valuation report"""
    
    context = practice_data.get_comprehensive_context()
    
    # Add provider breakdown if productivity data is available
    provider_context = ""
    if practice_data.productivity_data is not None and 'service_role' in practice_data.productivity_data.columns:
        df = practice_data.productivity_data
        latest_year = df['year'].max() if 'year' in df.columns else None
        
        if latest_year:
            latest_data = df[df['year'] == latest_year]
            
            # Get provider counts by role
            provider_counts = df.groupby('service_role')['provider'].nunique()
            
            # Get production by role for latest year
            role_production = latest_data.groupby('service_role')['production'].sum()
            
            # Get average production per provider by role
            provider_context = "\n\n=== PROVIDER BREAKDOWN ==="
            for role in df['service_role'].unique():
                role_data = latest_data[latest_data['service_role'] == role]
                providers = role_data['provider'].nunique()
                production = role_data['production'].sum()
                
                if providers > 0:
                    avg_prod = production / providers
                    provider_context += f"\n{role}s: {providers} providers, ${production:,.0f} total production, ${avg_prod:,.0f} average per provider"
            
            # Add industry benchmarks for context
            provider_context += """
            
=== INDUSTRY BENCHMARKS FOR CONTEXT ===
- General Dentist: $600,000-$800,000 annual production
- Specialist Dentist: $800,000-$1,200,000 annual production  
- Dental Hygienist: $150,000-$200,000 annual production
- Production per operatory: $150,000-$250,000 annually"""
    
    prompt = f"""You are an expert dental practice valuation analyst. Generate a comprehensive valuation report based on the following practice data.

{context}{provider_context}

IMPORTANT CONTEXT ABOUT THE DATA:
- The "Revenue" figure represents annual production (gross charges), not collections
- Production and collections are different - production is what's charged, collections are what's actually received
- When analyzing provider productivity, distinguish between dentists and hygienists as they have very different production expectations
- Collection rate = Collections / Production

Create a professional valuation report following this structure:

# Valuation Report: [Practice Name]

## Executive Summary

Write a professional summary that includes:
- Brief description of the practice based on available data
- Analysis of the current financial situation based on the metrics provided
- Your assessment of the practice's financial health
- Appropriate valuation methodology given the practice's financial condition
- Key findings and valuation conclusion

### Clinic Information
[Create a table with all available practice details from the data]

### Key Financial Indicators (Trailing Twelve Months)
[Create a comprehensive table of all available financial metrics]

### Valuation Summary
Based on your analysis, provide:
- Recommended valuation approach(es)
- Valuation range with justification
- Key factors influencing the valuation

## Practice Overview

### Provider Analysis
Analyze the provider mix and productivity:
- Breakdown of providers by role (dentists vs hygienists)
- Production analysis by provider type compared to industry benchmarks
- Assessment of provider productivity and capacity utilization
- Staffing efficiency analysis

### Operational Analysis
Analyze the available data to provide insights on:
- Patient volume and service utilization
- Revenue generation efficiency by provider type
- Collection effectiveness and revenue cycle management
- Operational strengths and concerns

### Financial Assessment
Based on the financial data provided, discuss:
- The practice's profitability position
- Key drivers of current financial performance
- Areas of financial strength or concern
- Comparison to industry benchmarks where relevant

## Financial Analysis

### Production and Collections Analysis
- Analyze gross production vs actual collections
- Calculate and assess the collection rate
- Identify revenue cycle management issues
- Quantify the financial impact of collection inefficiencies

### Trailing Twelve-Month Performance
[Provide detailed analysis of all financial metrics with professional commentary]

### Seller's Discretionary Earnings (SDE) Calculation

Create a detailed SDE table and provide analysis of:
- The components and their impact
- Whether the SDE is positive or negative and what this indicates
- The quality of earnings based on the add-backs

### SDE Add-back Justifications
[Explain each add-back category and assess its validity]

## Valuation Methodology

### Valuation Approach Selection
Based on the practice's financial condition:
- Determine and justify the most appropriate valuation method(s)
- Explain why certain methods may or may not be applicable
- Consider multiple approaches if relevant

### Detailed Valuation Analysis
Apply your selected methodology/methodologies and show:
- Calculations and assumptions
- Multiple ranges if using different approaches
- Sensitivity analysis where appropriate

## Growth Opportunities and Risk Assessment

### Opportunities for Enhancement
Based on your analysis of the data, identify:
- Provider productivity optimization opportunities
- Revenue cycle management improvements
- Operational efficiency gains
- Strategic growth possibilities

### Risk Factors
Identify and assess risks based on the financial data:
- Financial risks evident from the metrics
- Operational concerns
- Staffing and productivity risks
- Market or competitive risks

## Investment Considerations

Based on your comprehensive analysis, provide balanced assessment of:
- Key strengths of the practice
- Primary concerns or challenges
- Ideal buyer profile
- Critical success factors for a potential acquirer

## Conclusion

Provide a professional conclusion that:
- Summarizes your key findings
- States the recommended valuation range with confidence level
- Identifies the most critical factors affecting value
- Provides specific due diligence recommendations
- Offers your professional opinion on the investment opportunity

## Appendices (if relevant)
- Additional calculations or analyses
- Assumptions and limiting conditions
- Data sources and reliability assessment

IMPORTANT INSTRUCTIONS:
1. Analyze the data objectively and draw your own conclusions
2. Always distinguish between production (gross charges) and collections (actual receipts)
3. When analyzing provider productivity, compare dentists to dentist benchmarks and hygienists to hygienist benchmarks
4. If SDE is negative, explain the implications for valuation
5. If SDE is positive, apply appropriate multiples based on the practice's characteristics
6. Consider all relevant valuation approaches and justify your selection
7. Provide specific, data-driven insights rather than generic observations
8. Maintain professional tone while being direct about challenges
9. Scale your analysis complexity based on the data available
10. If certain expected data is missing, note this and explain the impact

Format everything in professional markdown with clear sections, tables, and bullet points."""
    
    return prompt

def generate_llm_prompt(question: str, practice_data: DentalPracticeData, category: str, provider_name: Optional[str] = None) -> str:
    """Generate a prompt based on question category and available data"""
    
    # Check if this is a valuation report request
    if category == 'report' or 'valuation report' in question.lower():
        return generate_valuation_report_prompt(practice_data)
    
    # For SDE or valuation specific questions
    if category in ['valuation', 'sde', 'financial']:
        prompt = f"""You are an expert dental practice valuation analyst. Answer the following question using the practice's financial and SDE data.

=== PRACTICE FINANCIAL DATA ===
{practice_data.get_comprehensive_context()}

Question: {question}

Provide a detailed answer that:
1. Directly addresses the question with specific numbers from the data
2. Explains the methodology or calculation if applicable
3. Provides context about what the numbers mean for practice valuation
4. Suggests relevant follow-up considerations

Use professional language suitable for practice owners and potential buyers."""
        return prompt
    
    # For productivity questions (existing functionality)
    prompt = "You are an expert dental practice analyst providing insights to practice owners and potential buyers.\n\n"
    
    # Add comprehensive context
    prompt += "=== COMPLETE PRACTICE DATA ===\n\n"
    prompt += practice_data.get_comprehensive_context()
    prompt += "\n\n"
    
    # Add productivity data if available and relevant
    if practice_data.productivity_data is not None and not practice_data.productivity_data.empty:
        if provider_name:
            prompt += f"=== SPECIFIC DATA FOR {provider_name.upper()} ===\n"
            provider_data = practice_data.productivity_data[
                practice_data.productivity_data['provider'] == provider_name
            ]
            if not provider_data.empty:
                prompt += "DETAILED RECORDS:\n"
                prompt += provider_data.to_string(index=False)
                prompt += "\n\n"
        else:
            prompt += "=== PRODUCTIVITY DATA ===\n"
            summary = practice_data.productivity_data.groupby(['provider', 'year']).agg({
                'production': 'sum',
                'collection': 'sum',
                'patients_seen': 'sum',
                'avg_visit_fee': 'mean'
            }).round(2)
            prompt += summary.to_string()
            prompt += "\n\n"
    
    # Add the question
    prompt += f"=== QUESTION ===\n{question}\n\n"
    
    # Instructions based on category
    if category == 'productivity':
        prompt += """=== INSTRUCTIONS ===
Focus on the specific metric asked about. Structure your response as:
1. **Direct Answer**: The specific metric with year-over-year changes
2. **Comparative Context**: How this compares to others or benchmarks
3. **Key Insights**: What the trend means for the practice
4. **Suggested Follow-ups**: Related questions to explore"""
    else:
        prompt += """=== INSTRUCTIONS ===
Provide a comprehensive answer that:
1. Uses specific data from the information provided
2. Includes relevant calculations or comparisons
3. Offers actionable insights
4. Maintains professional tone suitable for practice valuation context"""
    
    return prompt

def get_llm_response(prompt: str, max_retries: int = 3) -> str:
    """Get response from the LLM API with retry logic"""
    # Check if API key is set
    if OPENROUTER_API_KEY == "your-api-key-here":
        return "⚠️ API key not configured. Please set your OpenRouter API key in Cell 2."
    
    import time
    
    # Get fresh headers for each request
    headers = get_headers()
    
    for attempt in range(max_retries):
        try:
            print(f"Calling LLM API (attempt {attempt + 1}/{max_retries})...")
            
            response = requests.post(
                OPENROUTER_URL,
                headers=headers,
                json={
                    "model": MODEL,
                    "messages": [
                        {"role": "system", "content": "You are an expert dental practice valuation analyst with deep knowledge of practice financials, SDE calculations, and industry valuation multiples."},
                        {"role": "user", "content": prompt}
                    ],
                    "temperature": 0.3,  # Lower temperature for factual responses
                    "max_tokens": 4000,  # Increased for comprehensive reports
                    "stream": False
                },
                timeout=30
            )
            
            print(f"API Response Status: {response.status_code}")
            
            if response.status_code == 200:
                response_json = response.json()
                if 'choices' in response_json and len(response_json['choices']) > 0:
                    full_response = response_json['choices'][0]['message']['content']
                    if 'finish_reason' in response_json['choices'][0]:
                        if response_json['choices'][0]['finish_reason'] == 'length':
                            full_response += "\n\n[Note: Response was truncated due to length limit]"
                    return full_response
                else:
                    return f"Error: Unexpected API response format. Response: {response_json}"
            
            elif response.status_code == 401:
                return f"Authentication Error: Please check your API key in Cell 2."
            
            elif response.status_code == 402:
                return f"Insufficient Credits: Please add credits to your OpenRouter account."
            
            elif response.status_code == 429:
                wait_time = (attempt + 1) * 2
                print(f"Rate limited. Waiting {wait_time} seconds before retry...")
                time.sleep(wait_time)
                continue
                
            elif response.status_code == 502 or response.status_code == 503:
                wait_time = (attempt + 1) * 2
                print(f"Server error. Waiting {wait_time} seconds before retry...")
                time.sleep(wait_time)
                continue
                
            else:
                return f"Error: API returned status code {response.status_code}. Response: {response.text}"
        
        except requests.exceptions.Timeout:
            print(f"Request timed out on attempt {attempt + 1}")
            if attempt < max_retries - 1:
                time.sleep(2)
                continue
            else:
                return "Error: Request timed out after multiple attempts."
        
        except requests.exceptions.ConnectionError as e:
            print(f"Connection error on attempt {attempt + 1}: {str(e)}")
            if attempt < max_retries - 1:
                time.sleep(2)
                continue
            else:
                return f"Error: Connection failed after {max_retries} attempts."
        
        except Exception as e:
            print(f"Unexpected error on attempt {attempt + 1}: {str(e)}")
            if attempt < max_retries - 1:
                time.sleep(2)
                continue
            else:
                return f"Error calling LLM API: {str(e)}"
    
    return "Error: Failed to get LLM response after all retries."
```

```python
# ============================================================================
# CELL 5: Enhanced Visualization Generation Functions
# ============================================================================

def create_sde_trend_visualization(practice_data: DentalPracticeData) -> go.Figure:
    """Create SDE trend visualization over years"""
    if 'SDE_Historical' not in practice_data.data or not practice_data.data['SDE_Historical']:
        return None
    
    years = sorted(practice_data.data['SDE_Historical'].keys())
    sde_values = [practice_data.data['SDE_Historical'][year] for year in years]
    
    fig = go.Figure()
    
    # Add SDE trend line
    fig.add_trace(go.Scatter(
        x=years,
        y=sde_values,
        mode='lines+markers',
        name='SDE',
        line=dict(color='green', width=3),
        marker=dict(size=10),
        text=[f"${val:,.0f}" for val in sde_values],
        textposition="top center"
    ))
    
    # Add revenue trend if available
    if practice_data.productivity_data is not None and 'year' in practice_data.productivity_data.columns:
        revenue_by_year = practice_data.productivity_data.groupby('year')['production'].sum()
        fig.add_trace(go.Scatter(
            x=revenue_by_year.index,
            y=revenue_by_year.values,
            mode='lines+markers',
            name='Revenue',
            line=dict(color='blue', width=2, dash='dash'),
            marker=dict(size=8)
        ))
    
    fig.update_layout(
        title="SDE and Revenue Trends",
        xaxis_title="Year",
        yaxis_title="Amount ($)",
        template='plotly_white',
        height=400,
        hovermode='x unified'
    )
    
    return fig

def create_valuation_range_chart(practice_data: DentalPracticeData) -> go.Figure:
    """Create visualization of valuation ranges"""
    if 'Implied_Valuations' not in practice_data.data:
        return None
    
    valuations = practice_data.data['Implied_Valuations']
    
    # Prepare data for bar chart
    methods = []
    values = []
    colors = []
    
    if 'SDE_Low' in valuations:
        methods.extend(['SDE Low (3.5x)', 'SDE Mid (4.5x)', 'SDE High (5.5x)'])
        values.extend([valuations.get('SDE_Low', 0), 
                      valuations.get('SDE_Mid', 0), 
                      valuations.get('SDE_High', 0)])
        colors.extend(['lightcoral', 'yellow', 'lightgreen'])
    
    if 'EBITDA' in valuations:
        methods.append('EBITDA (4.5x)')
        values.append(valuations['EBITDA'])
        colors.append('lightblue')
    
    if 'Revenue' in valuations:
        methods.append('Revenue (0.8x)')
        values.append(valuations['Revenue'])
        colors.append('gray')
    
    fig = go.Figure()
    
    fig.add_trace(go.Bar(
        x=methods,
        y=values,
        marker_color=colors,
        text=[f"${val:,.0f}" for val in values],
        textposition='auto'
    ))
    
    fig.update_layout(
        title="Practice Valuation Range by Method",
        xaxis_title="Valuation Method",
        yaxis_title="Implied Value ($)",
        template='plotly_white',
        height=400
    )
    
    return fig

def create_sde_components_chart(practice_data: DentalPracticeData) -> go.Figure:
    """Create breakdown of SDE components"""
    if 'SDE_Components' not in practice_data.data:
        return None
    
    components = practice_data.data['SDE_Components']
    
    # Filter out zero values and prepare data
    labels = []
    values = []
    for key, value in components.items():
        if value > 0:
            labels.append(key.replace('_', ' '))
            values.append(value)
    
    if not labels:
        return None
    
    fig = go.Figure()
    
    fig.add_trace(go.Pie(
        labels=labels,
        values=values,
        hole=0.3,
        textinfo='label+percent',
        textposition='auto'
    ))
    
    fig.update_layout(
        title=f"SDE Components Breakdown (Total: ${sum(values):,.0f})",
        template='plotly_white',
        height=400
    )
    
    return fig

def create_financial_metrics_dashboard(practice_data: DentalPracticeData) -> go.Figure:
    """Create comprehensive financial metrics dashboard"""
    from plotly.subplots import make_subplots
    
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Revenue & SDE Trend', 'Margin Analysis', 
                       'SDE Components', 'Collection Performance'),
        specs=[[{'type': 'scatter'}, {'type': 'bar'}],
               [{'type': 'pie'}, {'type': 'indicator'}]]
    )
    
    # 1. Revenue & SDE Trend (if historical data available)
    if 'SDE_Historical' in practice_data.data:
        years = sorted(practice_data.data['SDE_Historical'].keys())
        sde_values = [practice_data.data['SDE_Historical'][year] for year in years]
        
        fig.add_trace(
            go.Scatter(x=years, y=sde_values, mode='lines+markers', name='SDE'),
            row=1, col=1
        )
    
    # 2. Margin Analysis
    margins = {
        'EBITDA Margin': practice_data.data.get('EBITDA_Margin', 0),
        'SDE Margin': practice_data.data.get('SDE_Margin', 0),
        'Collection Rate': practice_data.data.get('Collection_Rate', 0)
    }
    
    fig.add_trace(
        go.Bar(x=list(margins.keys()), y=list(margins.values()),
               text=[f"{v:.1f}%" for v in margins.values()],
               textposition='auto'),
        row=1, col=2
    )
    
    # 3. SDE Components
    if 'SDE_Components' in practice_data.data:
        components = {k: v for k, v in practice_data.data['SDE_Components'].items() if v > 0}
        if components:
            fig.add_trace(
                go.Pie(labels=[k.replace('_', ' ') for k in components.keys()], 
                       values=list(components.values()),
                       hole=0.3),
                row=2, col=1
            )
    
    # 4. Collection Rate Indicator
    collection_rate = practice_data.data.get('Collection_Rate', 0)
    fig.add_trace(
        go.Indicator(
            mode="gauge+number",
            value=collection_rate,
            title={'text': "Collection Rate %"},
            gauge={'axis': {'range': [None, 100]},
                   'bar': {'color': "darkgreen"},
                   'steps': [
                       {'range': [0, 85], 'color': "lightgray"},
                       {'range': [85, 95], 'color': "yellow"},
                       {'range': [95, 100], 'color': "lightgreen"}],
                   'threshold': {'line': {'color': "red", 'width': 4},
                                'thickness': 0.75, 'value': 95}}),
        row=2, col=2
    )
    
    fig.update_layout(
        title_text="Financial Performance Dashboard",
        showlegend=False,
        height=700,
        template='plotly_white'
    )
    
    return fig

def create_smart_visualization(practice_data: DentalPracticeData, question: str, answer: str = "") -> go.Figure:
    """Create visualization based on question context"""
    
    question_lower = question.lower()
    
    # Detect visualization type based on question
    if 'valuation' in question_lower or 'value' in question_lower or 'worth' in question_lower:
        return create_valuation_range_chart(practice_data)
    
    elif 'sde' in question_lower and 'trend' in question_lower:
        return create_sde_trend_visualization(practice_data)
    
    elif 'sde' in question_lower and ('component' in question_lower or 'breakdown' in question_lower):
        return create_sde_components_chart(practice_data)
    
    elif 'financial' in question_lower or 'dashboard' in question_lower:
        return create_financial_metrics_dashboard(practice_data)
    
    # For provider-specific questions (existing functionality)
    provider_name = detect_provider_in_question(question, practice_data)
    
    if provider_name and practice_data.productivity_data is not None:
        df = practice_data.productivity_data[practice_data.productivity_data['provider'] == provider_name]
        
        if df.empty:
            return None
        
        # Determine metric to visualize
        metric = None
        title = None
        
        if 'visit fee' in question_lower or 'avg' in question_lower:
            metric = 'avg_visit_fee'
            title = 'Average Visit Fee'
        elif 'production' in question_lower or 'productivity' in question_lower:
            metric = 'production'
            title = 'Production'
        elif 'collection' in question_lower:
            metric = 'collection'
            title = 'Collections'
        elif 'patient' in question_lower:
            metric = 'patients_seen'
            title = 'Patients Seen'
        
        if metric and metric in df.columns and 'year' in df.columns:
            # Aggregate by year
            if metric in ['production', 'collection', 'patients_seen']:
                yearly_data = df.groupby('year')[metric].sum().sort_index()
            else:
                yearly_data = df.groupby('year')[metric].mean().sort_index()
            
            fig = go.Figure()
            
            # Create appropriate chart type
            if 'trend' in question_lower or len(yearly_data) > 2:
                fig.add_trace(go.Scatter(
                    x=yearly_data.index,
                    y=yearly_data.values,
                    mode='lines+markers',
                    line=dict(width=3),
                    marker=dict(size=10),
                    text=[f"${val:,.2f}" if metric != 'patients_seen' else f"{val:,.0f}" 
                          for val in yearly_data.values],
                    textposition="top center"
                ))
            else:
                fig.add_trace(go.Bar(
                    x=yearly_data.index,
                    y=yearly_data.values,
                    text=[f"${val:,.2f}" if metric != 'patients_seen' else f"{val:,.0f}" 
                          for val in yearly_data.values],
                    textposition='auto'
                ))
            
            fig.update_layout(
                title=f"{title} - {provider_name}",
                xaxis_title="Year",
                yaxis_title=title,
                template='plotly_white',
                height=400
            )
            
            return fig
    
    # Default: try to create a relevant chart based on available data
    if 'SDE_Historical' in practice_data.data:
        return create_sde_trend_visualization(practice_data)
    
    return None

# Keep existing visualization functions for backward compatibility
def create_productivity_dashboard(practice_data: DentalPracticeData) -> go.Figure:
    """Create comprehensive productivity dashboard (existing function)"""
    if practice_data.productivity_data is None:
        return None
    
    df = practice_data.productivity_data
    
    # Check available columns
    has_year = 'year' in df.columns
    has_provider = 'provider' in df.columns
    has_production = 'production' in df.columns
    has_collection = 'collection' in df.columns
    has_patients = 'patients_seen' in df.columns
    has_visit_fee = 'avg_visit_fee' in df.columns
    
    from plotly.subplots import make_subplots
    
    fig = make_subplots(
        rows=2, cols=3,
        subplot_titles=('Production by Year', 'Collection Rate Trend', 
                       'Provider Performance', 'Patients Seen Trend',
                       'Production vs Collections', 'Average Visit Fee'),
        specs=[[{'type': 'bar'}, {'type': 'scatter'}, {'type': 'bar'}],
               [{'type': 'scatter'}, {'type': 'scatter'}, {'type': 'box'}]]
    )
    
    # 1. Production by Year
    if has_year and has_production:
        yearly_prod = df.groupby('year')['production'].sum()
        fig.add_trace(
            go.Bar(x=yearly_prod.index, y=yearly_prod.values, name='Production',
                   marker_color='lightblue'),
            row=1, col=1
        )
    
    # 2. Collection Rate Trend
    if has_year and has_production and has_collection:
        yearly_stats = df.groupby('year').agg({
            'production': 'sum',
            'collection': 'sum'
        })
        yearly_stats['collection_rate'] = (yearly_stats['collection'] / 
                                           yearly_stats['production'] * 100)
        fig.add_trace(
            go.Scatter(x=yearly_stats.index, y=yearly_stats['collection_rate'],
                      mode='lines+markers', name='Collection %',
                      line=dict(color='green', width=2)),
            row=1, col=2
        )
    
    # 3. Provider Performance
    if has_provider and has_production:
        provider_prod = df.groupby('provider')['production'].sum().nlargest(10)
        fig.add_trace(
            go.Bar(x=provider_prod.values, y=provider_prod.index,
                   orientation='h', name='Provider Production',
                   marker_color='coral'),
            row=1, col=3
        )
    
    # 4. Patients Seen Trend
    if has_year and has_patients:
        yearly_patients = df.groupby('year')['patients_seen'].sum()
        fig.add_trace(
            go.Scatter(x=yearly_patients.index, y=yearly_patients.values,
                      mode='lines+markers', name='Patients',
                      line=dict(color='purple', width=2)),
            row=2, col=1
        )
    
    # 5. Production vs Collections
    if has_production and has_collection:
        fig.add_trace(
            go.Scatter(x=df['production'], y=df['collection'],
                      mode='markers', name='Prod vs Coll',
                      marker=dict(size=5, color='orange')),
            row=2, col=2
        )
    
    # 6. Average Visit Fee Distribution
    if has_visit_fee:
        fig.add_trace(
            go.Box(y=df['avg_visit_fee'], name='Visit Fee',
                   marker_color='lightgreen'),
            row=2, col=3
        )
    
    fig.update_layout(
        title_text="Productivity Analysis Dashboard",
        showlegend=False,
        height=700,
        template='plotly_white'
    )
    
    return fig
```

```python
# ============================================================================
# CELL 6: Main Pipeline - Separate Report Generation and Q&A Functions
# ============================================================================

def generate_valuation_report(practice_data: DentalPracticeData, 
                             years: Optional[Union[int, List[int], str]] = None) -> Dict[str, Any]:
    """
    Generate a comprehensive valuation report with optional year filtering.
    
    Args:
        practice_data: DentalPracticeData object with all practice information
        years: Optional - can be:
            - None or "all": Generate report for all available years (default)
            - int: Generate report for a single year (e.g., 2022)
            - List[int]: Generate report for specific years (e.g., [2023, 2024])
            - "latest": Generate report for the most recent year
    
    Returns:
        Dictionary with report text and associated visualizations
    """
    
    # Handle year filtering if specified
    if years is not None and years != "all":
        print(f"Filtering data for year(s): {years}")
        
        # Reload data with year filter
        # Note: You'll need to store the file paths or pass them as parameters
        # For now, using hardcoded paths - you may want to make these configurable
        if isinstance(years, int):
            filter_years = [years]
        elif isinstance(years, str) and years == "latest":
            # Get the latest year from current data
            available_years = []
            if practice_data.pnl_sde_data is not None:
                year_col = 'Year' if 'Year' in practice_data.pnl_sde_data.columns else 'year'
                available_years = practice_data.pnl_sde_data[year_col].unique()
            filter_years = [max(available_years)]
        elif isinstance(years, list):
            filter_years = years
        else:
            filter_years = None
            
        # Create filtered practice_data
        practice_data = DentalPracticeData(
            productivity_csv="CSVs/productivity_2022_2025_all.csv",
            taxes_sde_csv="CSVs/Taxes_addback_analysis_all.csv",
            pnl_sde_csv="CSVs/P&L_2022_2025_addbacks.csv",
            filter_years=filter_years
        )
        
        report_period = str(years) if isinstance(years, int) else f"Years: {filter_years}"
    else:
        report_period = "All Available Years"
    
    print(f"Generating comprehensive valuation report for {report_period}...")
    
    # Generate the valuation report prompt
    prompt = generate_valuation_report_prompt(practice_data)
    
    # Get the full report from LLM
    report_text = get_llm_response(prompt)
    
    # Generate relevant visualizations for the report
    visualizations = {
        'sde_trend': create_sde_trend_visualization(practice_data),
        'valuation_range': create_valuation_range_chart(practice_data),
        'sde_components': create_sde_components_chart(practice_data),
        'financial_dashboard': create_financial_metrics_dashboard(practice_data)
    }
    
    # Remove None visualizations
    visualizations = {k: v for k, v in visualizations.items() if v is not None}
    
    return {
        'report': report_text,
        'visualizations': visualizations,
        'practice_data': {
            'total_sde': practice_data.data.get('Total_SDE', 0),
            'revenue': practice_data.data.get('Revenue_Trailing12', 0),
            'valuation_low': practice_data.data.get('Implied_Valuations', {}).get('SDE_Low', 0),
            'valuation_high': practice_data.data.get('Implied_Valuations', {}).get('SDE_High', 0),
        }
    }

def process_question_with_visualizations(
    question: str,
    practice_data: DentalPracticeData,
    show_visualization: bool = True
) -> Dict[str, Any]:
    """
    Process Q&A questions about the practice (existing functionality).
    This is for the Q&A tab, NOT for generating the valuation report.
    
    Args:
        question: User's question about the practice
        practice_data: DentalPracticeData object with all practice information
        show_visualization: Whether to generate and display visualizations
    
    Returns:
        Dictionary with answer and optional visualization
    """
    
    # Detect if a specific provider is mentioned
    provider_name = detect_provider_in_question(question, practice_data)
    if provider_name:
        print(f"Detected Provider: {provider_name}")
    
    # Determine general category for context
    category = categorize_question(question)
    print(f"Question Category: {category}")
    
    # IMPORTANT: Check if user is asking for a full report via question
    # If so, redirect them to use the report generation function
    if category == 'report' or 'valuation report' in question.lower() or 'generate report' in question.lower():
        return {
            'question': question,
            'provider': None,
            'answer': "To generate a comprehensive valuation report, please use the report generation function or the Report tab in the web application. This Q&A interface is for specific questions about the practice data.",
            'visualization': None,
            'redirect_to_report': True
        }
    
    # Generate prompt with actual data for Q&A
    prompt = generate_llm_prompt(question, practice_data, category, provider_name)
    
    # Get LLM response
    print("Getting LLM analysis...")
    answer = get_llm_response(prompt)
    
    # If LLM failed, provide fallback
    if "Error" in answer or "API key not configured" in answer:
        print("LLM unavailable, using data-based fallback...")
        answer = generate_data_based_answer(question, practice_data, provider_name)
    
    # Generate a simple visualization if relevant
    visualization = None
    if show_visualization:
        visualization = create_smart_visualization(practice_data, question, answer)
        if visualization:
            visualization.show()
    
    return {
        'question': question,
        'provider': provider_name,
        'category': category,
        'answer': answer,
        'visualization': visualization
    }

def generate_data_based_answer(question: str, practice_data: DentalPracticeData, provider_name: Optional[str]) -> str:
    """Generate a data-based answer when LLM is not available (fallback)"""
    answer = ""
    
    # Check if asking about financial/valuation metrics
    question_lower = question.lower()
    if any(term in question_lower for term in ['sde', 'valuation', 'worth', 'value', 'ebitda']):
        answer = f"""Based on the available data:

**Financial Metrics:**
- Total SDE: ${practice_data.data.get('Total_SDE', 0):,.0f}
- Revenue (TTM): ${practice_data.data.get('Revenue_Trailing12', 0):,.0f}
- EBITDA (TTM): ${practice_data.data.get('EBITDA_Trailing12', 0):,.0f}

**Valuation Range:**
- Low (3.5x SDE): ${practice_data.data.get('Implied_Valuations', {}).get('SDE_Low', 0):,.0f}
- Mid (4.5x SDE): ${practice_data.data.get('Implied_Valuations', {}).get('SDE_Mid', 0):,.0f}
- High (5.5x SDE): ${practice_data.data.get('Implied_Valuations', {}).get('SDE_High', 0):,.0f}

**SDE Components:**"""
        
        if 'SDE_Components' in practice_data.data:
            for component, value in practice_data.data['SDE_Components'].items():
                answer += f"\n- {component.replace('_', ' ')}: ${value:,.0f}"
        
        return answer
    
    # Handle provider-specific questions (existing functionality)
    if provider_name:
        has_productivity_data = False
        
        if practice_data.productivity_data is not None:
            provider_prod_data = practice_data.productivity_data[
                practice_data.productivity_data['provider'] == provider_name
            ]
            has_productivity_data = not provider_prod_data.empty
        
        if has_productivity_data:
            answer += f"Analysis for {provider_name}:\n\n"
            
            # Provide metrics based on question type
            if 'average visit fee' in question_lower or 'avg visit fee' in question_lower:
                if 'avg_visit_fee' in provider_prod_data.columns:
                    avg_fee_overall = provider_prod_data['avg_visit_fee'].mean()
                    by_year = provider_prod_data.groupby('year')['avg_visit_fee'].mean().sort_index()
                    
                    answer += f"**Average Visit Fee**: ${avg_fee_overall:.2f}\n\n"
                    answer += "**Breakdown by Year**:\n"
                    for year, fee in by_year.items():
                        answer += f"• {year}: ${fee:.2f}\n"
            
            elif 'productivity' in question_lower or 'production' in question_lower:
                if 'production' in provider_prod_data.columns:
                    total_prod = provider_prod_data['production'].sum()
                    by_year = provider_prod_data.groupby('year')['production'].sum().sort_index()
                    
                    answer += f"**Total Production**: ${total_prod:,.0f}\n\n"
                    answer += "**Year-by-Year Production**:\n"
                    for year, prod in by_year.items():
                        answer += f"• {year}: ${prod:,.0f}\n"
            
            else:
                # General summary
                latest_year = provider_prod_data['year'].max() if 'year' in provider_prod_data.columns else None
                if latest_year:
                    latest = provider_prod_data[provider_prod_data['year'] == latest_year]
                    answer += f"**{latest_year} Performance**:\n"
                    if 'production' in latest.columns:
                        answer += f"• Production: ${latest['production'].sum():,.0f}\n"
                    if 'collection' in latest.columns:
                        answer += f"• Collections: ${latest['collection'].sum():,.0f}\n"
                    if 'patients_seen' in latest.columns:
                        answer += f"• Patients Seen: {latest['patients_seen'].sum():,}\n"
                    if 'avg_visit_fee' in latest.columns:
                        answer += f"• Average Visit Fee: ${latest['avg_visit_fee'].mean():.2f}\n"
        else:
            answer = f"No data found for provider: {provider_name}"
    else:
        answer = "Please specify a provider name or ask about financial/valuation metrics for detailed analysis."
    
    return answer

def run_interactive_qa_session(practice_data: DentalPracticeData):
    """Run an interactive Q&A session (for testing purposes)"""
    print("\n=== Dental Practice Q&A System ===")
    print("Commands:")
    print("  'report' - Generate full valuation report")
    print("  'quit' - Exit the system")
    print("\nExample questions:")
    print("- What's the practice valuation?")
    print("- Show me the SDE breakdown")
    print("- What is the average visit fee for [Provider Name]?")
    print("- How's the productivity trending for [Provider Name]?")
    print("-" * 50)
    
    while True:
        question = input("\nYour question: ").strip()
        
        if question.lower() in ['quit', 'exit', 'q']:
            print("Thank you for using the Q&A system!")
            break
        
        if question.lower() == 'report':
            print("\nGenerating valuation report...")
            report_data = generate_valuation_report(practice_data)
            print("\n" + "="*60)
            print("VALUATION REPORT")
            print("="*60)
            print(report_data['report'])
            print("="*60)
            print(f"\nGenerated {len(report_data['visualizations'])} visualizations")
            continue
        
        if not question:
            continue
        
        try:
            result = process_question_with_visualizations(question, practice_data)
            print(f"\n{'='*50}")
            print(f"Category: {result.get('category', 'N/A')}")
            if result.get('provider'):
                print(f"Provider: {result['provider']}")
            print(f"{'='*50}")
            print(f"\nAnswer:\n{result['answer']}")
            print(f"{'='*50}")
            
        except Exception as e:
            print(f"Error processing question: {e}")

# ============================================================================
# Example Usage - How to integrate with web application
# ============================================================================

if __name__ == "__main__":
    # Load practice data with the new CSV files
    practice_data = DentalPracticeData(
        productivity_csv="CSVs/productivity_2022_2025_all.csv",
        taxes_sde_csv="CSVs/Taxes_addback_analysis_all.csv",
        pnl_sde_csv="CSVs/P&L_2022_2025_addbacks.csv"
    )
    
    print("Data loaded successfully!")
    print(f"Total SDE: ${practice_data.data.get('Total_SDE', 0):,.0f}")
    print(f"Revenue (TTM): ${practice_data.data.get('Revenue_Trailing12', 0):,.0f}")
    
    # Example 1: Generate valuation report (for Report tab in web app)
    # Uncomment to test:
    # report_data = generate_valuation_report(practice_data)
    # print(report_data['report'][:2000])  # Print first 2000 chars
    
    # Example 2: Process Q&A (for Questions tab in web app)
    # Uncomment to test:
    # question = "What is the average visit fee for Emily Hawkins, RDH?"
    # result = process_question_with_visualizations(question, practice_data)
    # print(result['answer'])
```
